# NLU - Natural Language Understanding 

Natural Language Understanding (NLU) is a subfield of Natural Language Processing that provides machines with the ability to interpret and extract meaning from human language. It enables systems to understand the intent behind linguistic inputs. NLU serves as the foundation for a wide range of language-driven applications including chatbots, virtual assistants and content moderation systems

As machines operate in binary code, the gap between language and machine-readable information requires language understanding. NLU helps this cause by applying statistical methods to analyze syntax, semantics and dependencies in text.

**Example**

- Siri/Alexa/Google Assistant: "Set a timer for 10 minutes" → NLU extracts intent (SetTimer) and entity (10 minutes)
- Customer service chatbots: "I want to cancel my order #12345" → intent = CancelOrder, entity = order ID 12345
- Banking assistants: "What's my checking account balance?" → intent = CheckBalance, entity = checking account

## What is Text Processing ?

Text processing refers to the set of techniques used to manipulate, analyze, clean, and transform raw text into a format that's easier for computers (or humans) to work with. It's the foundational step that happens before deeper tasks like NLU, sentiment analysis, or machine translation — think of it as preparing raw ingredients before cooking.

Text processing can be rule-based (simple pattern operations) or feed into NLP/ML pipelines (structured preparation for models).

**Why it is needed** 

Raw text is messy — full of typos, inconsistent casing, punctuation, irrelevant words, and different word forms. Text processing cleans and structures it so models/algorithms can extract meaningful patterns instead of getting confused by surface-level noise.

### Step 3 Tokenization

Tokenization is the process of breaking down a piece of text into smaller units called tokens - these can be sentences, words, subwords, or even characters, depending on the level of granualarity needed. It usually the very first step in any NLP, NLU pipeline, since computers need text split into descrete, managable pieces before they can analyze it. 

**Types of Tokenization**

- **Sentence Tokenization**

Splits a larger body of text (a paragraph or document) into individual sentences. This is useful when you need to analyze or process text sentence-by-sentence (e.g., summarization, translation).

Example: 

Input
"NLU is a part of NLP. It focuses on understanding meaning. Many applications use it today."

Output
["NLU is a part of NLP.",
 "It focuses on understanding meaning.",
 "Many applications use it today."]

 Sentence tokenization isn't as simple as just splitting on periods — it has to handle tricky cases like:

"Dr. Smith went to the U.S. yesterday." → should be one sentence, not split at "Dr." or "U.S."

- **Word Tokenization**

Splits a sentence into individual words (and often punctuation marks as separate tokens too). This is the most common form of tokenization used before tasks like POS tagging, NER, or feeding text into embeddings/models.

Example: 

Input - "NLU is powerful."

Output - ["NLU", "is", "powerful", "."]

**Why Tokenization Matter**

- It's the foundation for every downstream NLP task — POS tagging, NER, parsing, sentiment analysis all operate on tokens
- Poor tokenization (e.g., mishandling contractions, abbreviations, or punctuation) can cascade into errors in every later step

#### Step 5 Stemming 

Stemming is a text preprocessing technique in NLP that reduces words to their root or base form by removing prefixes and suffixes. It helps simplify and standardize text, making text analysis and processing more efficient

- Simplifies words into a common root form
- Improves text processing and analysis efficiency
- Commonly used in text classification and information retrieval
- Helps reduce redundancy in text data
- May sometimes reduce readability or produce inaccurate root words

**Types of Stemmer in NTLK**

1. **Porter's Stemmer**: It is one of the most widely used stemming algorithms in NLP. It removes common suffixes from English words using a set of predefined rules to produce root forms.

Example : 'agreed' → 'agree'
Rule: If the word has a suffix EED (with at least one vowel and consonant) remove the suffix and change it to EE.

2. **Snowball Stemmer**: Snowball Stemmer is an improved version of Porter’s Stemmer, also known as Porter2. It is faster, more aggressive, and supports multiple languages for stemming tasks.

- Improved and faster version of Porter’s Stemmer
- Removes suffixes more effectively
- Supports multiple languages
- Commonly used for multilingual text processing
- Produces more consistent stemming results

Example- 

'running' → 'run'

'quickly' → 'quick'

3. Lancaster Stemmer 

Lancaster Stemmer is a fast and aggressive stemming algorithm that reduces words to very short root forms using iterative rule-based processing.

- Faster and more aggressive than many other stemmers
- Applies rules repeatedly to reduce words
- Can produce very short or distorted stems
- Useful for applications requiring fast text processing
- May reduce accuracy due to over-stemming

Example

'running' → 'run'

'happily' → 'happy'

4. Regexp Stemmer

Regexp Stemmer is a flexible stemming algorithm that uses regular expressions (regex) to remove word suffixes based on custom rules.

- Allows custom stemming rules using regex
- Useful for domain-specific text processing
- Provides flexible and rule-based stemming
- Requires manual rule creation
- Can be slower for large datasets

Example

'running' → 'runn'
Custom rule: r'ing$' removes the suffix ing.

5. Krovetz Stemmer 

Krovetz Stemmer is a linguistically aware stemming algorithm developed by Robert Krovetz. It focuses on preserving the actual meaning of words while converting them into their root forms.

- Preserves linguistic meaning more accurately
- Handles plural and tense conversions effectively
- Produces more meaningful root words
- Slower compared to many other stemmers
- May be less efficient for very large datasets

Example

'children' → 'child'
'running' → 'run

**Advantage of Stemmer**

- Normalizes different word forms into a common root
- Reduces text dimensionality and improves efficiency
- Enhances search and information retrieval performance
- Simplifies processing of large text datasets
- Improves machine learning and text analysis tasks

**Limitation of Stemmer**

- Over-stemming may reduce words too aggressively and change their meaning
- Under-stemming may fail to group related words into the same root form
- Ignores context and semantic meaning of words
- Can affect accuracy in tasks like sentiment analysis
- Different stemmers may produce different results for the same word

#### Lemmatization 

Lemmatization is a text preprocessing technique in Natural Language Processing (NLP) that converts words into their base or dictionary form called a lemma. Unlike stemming, it considers the meaning and part of speech of words, making the output more accurate and meaningful.

**Techniques of lemmatization**

1. Rule Based Lemmatization

In rule-based lemmatization, predefined rules are applied to a word to remove suffixes and get the root form. This approach works well for regular words but may not handle irregularities well.

For example:

Rule: For regular verbs ending in "-ed," remove the "-ed" suffix.

Example: "walked" -> "walk"

While this method is simple and interpretable, it doesn't account for irregular word forms like "better" which should be lemmatized to "good".

2. Dictionary-Based Lemmatization

It uses a predefined dictionary or lexicon such as WordNet to look up the base form of a word. This method is more accurate than rule-based lemmatization because it accounts for exceptions and irregular words.

For example:

'running' -> 'run'
'better' -> 'good'
'went' -> 'go
"I was running to become a better athlete and then I went home," -> "I was run to become a good athlete and then I go home."

By using dictionaries like WordNet this method can handle a range of words effectively, especially in languages with well-established dictionaries.

3. Machine Learning-Based Lemmatization

It uses algorithms trained on large datasets to automatically identify the base form of words. This approach is highly flexible and can handle irregular words and linguistic nuances better than the rule-based and dictionary-based methods.

For example:

A trained model may deduce that “went” corresponds to “go” even though the suffix removal rule doesn’t apply. Similarly, for 'happier' the model deduces 'happy' as the lemma. 

Machine learning-based lemmatizers are more adaptive and can generalize across different word forms which makes them ideal for complex tasks involving diverse vocabularies.

**Differences**


| Aspect | Stemming | Lemmatization |
|---|---|---|
| **Definition** | Reduces a word to its root form by chopping off prefixes/suffixes | Reduces a word to its dictionary base form (lemma) using vocabulary and grammar rules |
| **Approach** | Rule-based, heuristic (e.g., removing "-ing", "-es", "-s") | Uses morphological analysis and a lexicon (e.g., WordNet) |
| **Output** | May not be a real word | Always a valid dictionary word |
| **Speed** | Faster (simple rule-based cutting) | Slower (requires vocabulary lookup + POS tagging) |
| **Accuracy** | Lower — can produce incorrect roots | Higher — context-aware and linguistically accurate |
| **Context Awareness** | No — ignores part of speech | Yes — considers part of speech (noun, verb, etc.) |
| **Example: "running"** | `run` | `run` |
| **Example: "studies"** | `studi` | `study` |
| **Example: "better"** | `better` (no change) | `good` |
| **Common Libraries** | `PorterStemmer`, `SnowballStemmer` (NLTK) | `WordNetLemmatizer` (NLTK), `spaCy` |

### Key Takeaway
- Use **stemming** when speed matters more than precision (e.g., search engines, information retrieval).
- Use **lemmatization** when meaning and grammatical correctness matter (e.g., chatbots, sentiment analysis).

#### POS Tagging 

Parts of Speech (PoS) tagging is a fundamental task in Natural Language Processing (NLP) where each word in a sentence is assigned a grammatical category such as noun, verb, adjective or adverb.

- This process help machines to understand the structure and meaning of sentences by identifying the roles of words and their relationships.
- Analyzes the context of words to determine their correct part of speech
- Improves text analysis, parsing, machine translation, and information extraction
- Widely used in text classification, sentiment analysis, chatbots, and search engine

**Common POS Tags**

| Tag | Meaning | Example |
|---|---|---|
| `NN` | Noun, singular | dog, city |
| `NNS` | Noun, plural | dogs, cities |
| `VB` | Verb, base form | run, eat |
| `VBD` | Verb, past tense | ran, ate |
| `JJ` | Adjective | happy, blue |
| `RB` | Adverb | quickly, well |
| `PRP` | Pronoun | he, she, it |
| `IN` | Preposition | in, on, at |
| `DT` | Determiner | the, a, an |

**Example**

Sentence: The quick brown fox jumps over the lazy dog

| Word | POS Tag | Meaning |
|---|---|---|
| The | DT | Determiner |
| quick | JJ | Adjective |
| brown | JJ | Adjective |
| fox | NN | Noun |
| jumps | VBZ | Verb, 3rd person singular present |
| over | IN | Preposition |
| the | DT | Determiner |
| lazy | JJ | Adjective |
| dog | NN | Noun |

**Types of POS Tagging**

<u>1. Rule-Based POS Tagging</u>

- Uses a set of hand-written **linguistic rules** to assign tags
- Relies on dictionary lookup + contextual rules (e.g., "if previous word is a determiner, tag current word as noun")
- One of the oldest approaches (e.g., **Brill Tagger**)
- **Pros:** Simple, interpretable
- **Cons:** Requires manual rule creation, doesn't generalize well

<u> 2. Stochastic / Probabilistic POS Tagging</u>

- Uses **statistical models** to assign the most probable tag based on frequency/probability
- Common techniques:
  - **Hidden Markov Model (HMM)** — uses transition & emission probabilities
  - **N-gram Tagger** — assigns tags based on the previous N words/tags
- **Pros:** Data-driven, works well with sufficient training data
- **Cons:** Needs large annotated corpus

 <u>3.Transformation-Based Tagging (Brill Tagging)</u>

- Combines rule-based and stochastic approaches
- Starts with an initial tag (often most frequent tag) and applies **transformation rules** learned from training data to correct errors
- **Pros:** More accurate than pure rule-based, still interpretable
- **Cons:** Slower training process

<u> 4. Deep Learning / Neural POS Tagging</u>

- Uses neural networks such as:
  - **RNN / LSTM / BiLSTM**
  - **CRF (Conditional Random Fields)** — often combined with BiLSTM (BiLSTM-CRF)
  - **Transformer-based models** (e.g., BERT, spaCy's neural pipeline)
- **Pros:** Highest accuracy, captures context and long-range dependencies
- **Cons:** Requires more computation and training data



#### Chunking (Shallow Parsing)

**Chunking**, also known as **shallow parsing**, is the process of grouping words into meaningful phrases (chunks) — such as noun phrases (NP), verb phrases (VP), or prepositional phrases (PP) — based on their POS tags, without building a full parse tree.

It sits between **POS tagging** (word-level labels) and **full parsing** (complete grammatical structure), extracting useful phrase-level chunks quickly.

##### Why Chunking is Important
- Extracts meaningful phrases (e.g., noun phrases for entity extraction)
- Faster and simpler than full syntactic parsing
- Useful for **information extraction**, **named entity recognition (NER)**, and **question answering**

## Named Entity Recognition (NER) and BIO Tagging

### What is NER?

**Named Entity Recognition (NER)** is an NLP technique that identifies and classifies **named entities** in text into predefined categories such as person names, organizations, locations, dates, monetary values, etc.

It helps machines understand **"who," "what," "where," and "when"** in a piece of text.

### Common Entity Types

| Entity Type | Tag | Example |
|---|---|---|
| Person | PERSON | Elon Musk, Marie Curie |
| Organization | ORG | Google, United Nations |
| Location | GPE/LOC | Paris, Mount Everest |
| Date | DATE | January 5, 2024 |
| Time | TIME | 10:30 AM |
| Money | MONEY | $500, ₹1000 |
| Percentage | PERCENT | 25% |

### Why NER is Important
- Powers **information extraction** from unstructured text
- Used in **search engines**, **chatbots**, **resume parsing**, **question answering**
- Helps in building **knowledge graphs** and **document summarization**

---

### What is BIO Tagging?

**BIO Tagging** (also called **IOB tagging**) is a labeling scheme used to represent entity boundaries at the **token level**, so each word in a sentence gets tagged based on its position within an entity.

### BIO Tag Scheme

| Tag | Meaning |
|---|---|
| `B-<TYPE>` | **Beginning** of an entity |
| `I-<TYPE>` | **Inside** an entity (continuation) |
| `O` | **Outside** — not part of any entity |

### Example

**Sentence:** `"Barack Obama was born in Hawaii"`

| Word | BIO Tag |
|---|---|
| Barack | B-PER |
| Obama | I-PER |
| was | O |
| born | O |
| in | O |
| Hawaii | B-LOC |

Here:
- `Barack Obama` → tagged as a single **PERSON** entity (`B-PER` + `I-PER`)
- `Hawaii` → tagged as a **LOCATION** entity (`B-LOC`)
- All other words are `O` (outside any entity)

### Why BIO Tagging is Used
- Allows **multi-word entities** to be represented clearly at the token level
- Standard input/output format for training **sequence labeling models** (HMM, CRF, BiLSTM-CRF, BERT-NER)
- Distinguishes between two adjacent entities of the same type (e.g., two separate PERSON names next to each other)

### Variants of BIO Tagging

| Scheme | Description |
|---|---|
| **IO** | Only `I` and `O` (no distinct beginning tag) — cannot separate adjacent entities |
| **BIO / IOB** | `B`, `I`, `O` — most common scheme |
| **BIOES** | Adds `E` (End) and `S` (Single) for more precise boundaries |



## One-Hot Encoding

**One-Hot Encoding** is a technique used to convert **categorical data** (including words/tokens in NLP) into a **numerical vector format** that machine learning models can understand.

Each category/word is represented as a **binary vector** where only **one position is "hot" (1)** and all other positions are **"cold" (0)**.

### Why One-Hot Encoding is Needed
- ML/DL models work with **numbers**, not text or labels
- Prevents models from assuming a **false ordinal relationship** between categories (e.g., "Red=1, Green=2, Blue=3" would wrongly imply Blue > Green > Red)
- Simple and widely used for representing categorical features and vocabulary in early NLP pipelines

### How It Works
1. Identify all **unique categories/words** → build a vocabulary
2. Assign each category an **index**
3. Create a vector of length = vocabulary size, with `1` at that category's index and `0` elsewhere

## TF-IDF (Term Frequency – Inverse Document Frequency)

**TF-IDF** is a statistical measure used to evaluate how **important a word** is to a document within a collection (corpus) of documents. It balances how frequently a word appears in a document against how common that word is across all documents.

The core idea: words that appear **often in one document but rarely across others** are more meaningful/important than words that appear everywhere (like "the", "is", "and").

### Formula

**TF-IDF = TF × IDF**

#### 1. Term Frequency (TF)
Measures how frequently a term appears in a document.

$$
TF(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d}
$$

#### 2. Inverse Document Frequency (IDF)
Measures how rare/important a term is across all documents.

$$
IDF(t) = \log \left( \frac{\text{Total number of documents}}{\text{Number of documents containing term } t} \right)
$$

#### 3. TF-IDF Score

$$
TFIDF(t, d) = TF(t, d) \times IDF(t)
$$

### Intuition
- **High TF-IDF** → word is frequent in this document but rare in others → **important/distinctive**
- **Low TF-IDF** → word is either rare in this document or common across all documents → **less important**

### Example

**Corpus (3 documents):**
1. "the cat sat on the mat"
2. "the dog sat on the log"
3. "cats and dogs are great pets"

The word **"the"** appears in almost every document → **low IDF** → low importance despite high frequency.

The word **"cat"** appears in only one document → **high IDF** → more important/distinctive for that

## N-Gram Models (Unigram, Bigram, Trigram, Skip-gram)

An **N-gram** is a contiguous sequence of **N items (words or characters)** from a given text. N-gram models are used to capture **local word order and context**, which helps in tasks like language modeling, text prediction, and feature extraction.

### General Formula
For a sentence with words `w1, w2, w3, ..., wn`, an N-gram model predicts/represents sequences of N consecutive words.

$$
P(w_1, w_2, ..., w_n) \approx \prod_{i=1}^{n} P(w_i \mid w_{i-N+1}, ..., w_{i-1})
$$

---

### 1. Unigram (N=1)
- Considers **single words** independently, with **no context** from surrounding words
- Simplest form — basically the same as Bag of Words at the word level

**Sentence:** `"I love NLP"`

**Unigrams:** `["I"], ["love"], ["NLP"]`

- **Pros:** Simple, fast
- **Cons:** No context — "not good" and "good" may seem similar since word order is ignored

---

### 2. Bigram (N=2)
- Considers **pairs of consecutive words**
- Captures some local context (e.g., which word follows another)

**Sentence:** `"I love NLP"`

**Bigrams:** `["I love"], ["love NLP"]`

---

### 3. Trigram (N=3)
- Considers **three consecutive words**
- Captures more context than bigrams, useful for better language modeling

**Sentence:** `"I love NLP a lot"`

**Trigrams:** `["I love NLP"], ["love NLP a"], ["NLP a lot"]`

- **Pros:** Better context capture than unigram/bigram
- **Cons:** Sparsity increases — many trigrams may never appear in training data (data sparsity problem)

---

### Comparison: Unigram vs Bigram vs Trigram

| N-gram | N | Example (from "I love NLP a lot") | Context Captured |
|---|---|---|---|
| Unigram | 1 | I, love, NLP, a, lot | None |
| Bigram | 2 | I love, love NLP, NLP a, a

## Continuous Bag of Words (CBOW)

**Continuous Bag of Words (CBOW)** is one of the two main architectures used in **Word2Vec** (the other being **Skip-gram**) to learn dense word embeddings. CBOW predicts a **target/center word** based on its surrounding **context words**.

### Key Idea
Given the **context words** within a window, CBOW tries to predict the **missing center word** — essentially the opposite task of Skip-gram.

> **CBOW:** Context → predicts → Target word
> **Skip-gram:** Target word → predicts → Context

### Example

**Sentence:** `"I love learning NLP deeply"`, window size = 2, target word = `"learning"`

**Context words (input):** `["I", "love", "NLP", "deeply"]`
**Target word (output to predict):** `"learning"`

So CBOW takes the context words `["I", "love", "NLP", "deeply"]` as input and tries to predict `"learning"` as the output.

### CBOW Architecture

Input: Context words (one-hot vectors) → I, love, NLP, deeply
↓
Average/Sum of embeddings
↓
Hidden Layer (embedding layer)
↓
Output: Predicted probability distribution over vocabulary
↓
Target word → "learning"


Unlike Skip-gram (which has one input word), CBOW takes **multiple context words as input**, combines (typically averages) their embeddings, and predicts a **single target word** as output.

### Training Pairs Example

**Sentence:** `"I love learning NLP deeply"`, window size = 2

| Target Word | Context Words (Input) |
|---|---|
| I | love, learning |
| love | I, learning, NLP |
| learning | I, love, NLP, deeply |
| NLP | love, learning, deeply |
| deeply | learning, NLP |

---

### CBOW vs Skip-gram

| Aspect | CBOW | Skip-gram |
|---|---|---|
| Direction | Context → Target | Target → Context |
| Input | Multiple context words | Single target word |
| Output | Single target word | Multiple context words |
| Training Speed | **Faster** (fewer training pairs per example) | Slower |
| Performance on Rare Words | Weaker | **Better** — treats each context word as a new example |
| Performance on Frequent Words | **Better** | Good |
| Dataset Size Suitability | Works well with **large datasets** | Works well with **small datasets** |

### Advantages of CBOW
- Faster to train compared to Skip-gram
- Works well when the corpus is large
- Smooths over noise since it averages multiple context words

### Disadvantages of CBOW
- Less effective at learning good representations for **rare words**
- Loses word order information (context words are simply averaged)

---

### Code Example — CBOW using Word2Vec (Gensim)
```python
from gensim.models import Word2Vec

sentences = [["I", "love", "learning", "NLP", "deeply"],
             ["NLP", "is", "an", "exciting", "field"]]

# sg=0 enables CBOW (sg=1 would be Skip-gram)
model = Word2Vec(sentences, vector_size=50, window=2, min_count=1, sg=0)

# Get vector for a word
print(model.wv["learning"])

# Find similar words
print(model.wv.most_similar("NLP"))
```

### Visual Summary

[I, love, NLP, deeply] ---> predicts ---> "learning"

Skip-gram:
"learning" ---> predicts ---> [I, love, NLP, deeply]


# RNN, LSTM, and GRU

**RNN, LSTM,** and **GRU** are neural network architectures designed to handle **sequential data** (text, speech, time series), where the order of inputs matters and previous context influences future predictions.

---

## 1. RNN (Recurrent Neural Network)

A **Recurrent Neural Network** processes sequences one element at a time while maintaining a **hidden state** that carries information from previous time steps, giving it a form of "memory."

**Formula:**
$$
h_t = \tanh(W_h h_{t-1} + W_x x_t + b)
$$

**Architecture:**

x1 → [RNN Cell] → h1 → [RNN Cell] → h2 → [RNN Cell] → h3 → .


**Pros:** Simple, fewer parameters, works well for short sequences
**Cons:** Suffers from **vanishing/exploding gradients**, poor long-term memory


## 2. LSTM (Long Short-Term Memory)

**LSTM** is an advanced RNN that solves the vanishing gradient problem using a **cell state** and **gates** that control what information to keep, forget, or output.

| Gate | Purpose |
|---|---|
| **Forget Gate** | Decides what to discard from cell state |
| **Input Gate** | Decides what new info to add to cell state |
| **Output Gate** | Decides what to output as hidden state |

**Pros:** Handles **long-term dependencies** well, solves vanishing gradient problem
**Cons:** More complex, slower, more parameters

---

## 3. GRU (Gated Recurrent Unit)

**GRU** is a simplified LSTM variant that merges the forget/input gates into a single **update gate**, and combines the cell state and hidden state into one.

| Gate | Purpose |
|---|---|
| **Update Gate** | Decides how much past info to keep |
| **Reset Gate** | Decides how much past info to forget |

**Pros:** Fewer parameters, faster training, comparable performance to LSTM
**Cons:** Slightly weaker on very long-term dependencies

## Comparison Table

| Aspect | RNN | LSTM | GRU |
|---|---|---|---|
| Memory Mechanism | Simple hidden state | Cell state + hidden state | Hidden state only |
| Gates | None | 3 (Forget, Input, Output) | 2 (Update, Reset) |
| Long-term Dependency | Poor | **Excellent** | Good |
| Training Speed | Fastest | Slowest | Faster than LSTM |
| Parameters | Fewest | Most | Moderate |
| Vanishing Gradient | Yes (major issue) | Solved | Solved |
| Best Use Case | Short sequences | Complex, long sequences | Balance of speed & performance |

## Code Example — RNN, LSTM, GRU using Keras/TensorFlow

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense, Embedding

vocab_size = 5000
embedding_dim = 64
max_length = 100

# Simple RNN Model
rnn_model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=max_length),
    SimpleRNN(64),
    Dense(1, activation='sigmoid')
])

# LSTM Model
lstm_model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=max_length),
    LSTM(64),
    Dense(1, activation='sigmoid')
])

# GRU Model
gru_model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=max_length),
    GRU(64),
    Dense(1, activation='sigmoid')
])

lstm_model.summary()
```

## Key Takeaway
- **RNN** → simplest, but suffers from vanishing gradients, poor at long-term memory
- **LSTM** → solves long-term dependency issues using gates and a separate cell state, but is computationally heavier
- **GRU** → a lighter, faster alternative to LSTM with comparable performance, using fewer gates and no separate cell state

In practice: use **GRU** when speed/efficiency matters, and **LSTM** when maximum accuracy on long sequences is the priority. Modern NLP has largely moved toward **Transformer-based models** (BERT, GPT) which outperform RNN/LSTM/GRU on most tasks.
