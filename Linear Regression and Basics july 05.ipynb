# Machine Learning 

### What is Machine Learning ?

Machine learning is a branch of Artificial Intelligence that focuses on developing models and algorithms that let computers learn from data without being explicitly programmed for every task. In simple words, ML teaches systems to think and understand like humans by learning from the data.

Example of Machine Learning :
- Imagine you want a computer to recognize cats in photos.

- Traditional programming: You would have to write rules like "cats have pointed ears, whiskers, and four legs."
- Machine learning: You show the computer thousands of labeled pictures of cats and non-cats. It learns the patterns on its own and can then identify cats in new images.

#### Lifecycle of Machine Learning


#### Machine learning model types

- Supervised Learning : Trains models on labeled data to predict or classify new, unseen data.(e.g: Predicting house prices based on size, location, and number of bedrooms.,Predicting whether an email is spam or not. predicting customer churn)

- Unsupervised Learning : Finds patterns or groups in unlabeled data, like clustering or dimensionality reduction. (e.g.: Customer segmentation in marketing, Grouping similar products, Detecing unsual behaviour (anomaly detction), Organizing new artciles by topic)

- Reinforcement Learning: Learns through trial and error to maximize rewards, ideal for decision-making tasks.(e.g.: Self driving class , roborts learning to walk , AI playing chess or video games)

- Self Survised learning: It is often considered as a  subset of unsupervised learning, but it has grown into its own field due to its success in training large-scale models. It generates its own labels from the data, without any manual labeling.

- Semi Supervised learning: This approach combines a small amount of labeled data with a large amount of unlabeled data. It’s useful when labeling data is expensive or time-consuming. (e.g.: Medical image classification, Face recognition system, Speech recongition, Image classification with only a few labeled images)

# Supervised Learning

Supervised learning is a type of machine learning in which the model is trained using labeled data. Labeled data means that each input has a known or correct output. The model learns the relationship between the input and the output and uses that knowledge to predict the output for new, unseen data.

In simple words, supervised learning is like learning with a teacher. The teacher provides the correct answers during training, and the model learns from those examples

**Its main features**

- Each input has a known output
- Adjusts itself to reduce prediction errors
- Make accurate predictions on new data
- For example it recognizing handwritten digits from trained data

**Types of supervised learning**

- Classification: Where output is a categorical variable (eg: spam vs non spam emails, yes or no)
- Regression: Where the output is containous variable(e.g, predicting house and stock prices)

### Working of supervised Machine learning 

1. Collect Labeled data:
- Gather a dataset where each input has a known correct output (label).
- Example: Images of handwritten digits with their actual numbers as labels.

2. Split the dataset 
- Divide the data into training data (about 80%) and testing data (about 20%).
- The model will learn from the training data and be evaluated on the testing data.

3. Train the model 
- Feed the training data (inputs and their labels) to a suitable supervised learning algorithm (like Decision Trees, SVM or Linear Regression).
- The model tries to find patterns that map inputs to correct outputs.

4. Validate and Test the model
- Evaluate the model using testing data it has never seen before.
- The model predicts outputs and these predictions are compared with the actual labels to calculate accuracy or error.

5. Deploy and predict on New data
- Once the model performs well, it can be used to predict outputs for completely new, unseen data

### Classification Supervised learning

Classification is a supervised machine learning technique used to predict labels or categories from input data. It assigns each data point to a predefined class based on learned patterns.

- Predict categories: Determines the class of new data points.
- Uses labeled data: Trained on datasets where the correct class is known.
- Common examples: Spam vs non spam emails, diseased vs. healthy patients.

For example: A classification model might be trained on dataset of images labeled as either dogs or cats and it can be used to predict the class of new and unseen images as dogs or cats based on their features such as colour, texture or shape.

**Types of classification**

Classification in machine learning involves sorting data into categories based on their features or characteristics. The type of classification problem depends on how many classes exist and how the categories are structured.



1. Binary Classification: Predicts one of exactly two outcomes (e.g., "Spam" vs. "Not Spam", "Pass" vs. "Fail")
2. Multiclass Classification: Predicts one class out of three or more mutually exclusive categories (e.g., classifying a photo as either a "Cat", "Dog", or "Bird")
3. Multilabel Classification: Assigns multiple, non-exclusive labels to a single data point simultaneously (e.g., tagging a movie as both "Action" and "Sci-Fi").
4. Imbalanced Classification: A specialized scenario where the number of examples in one class vastly outnumbers the other, requiring adjusted metrics or resampling techniques (e.g., credit card fraud detection).

## Linear Regression 

Linear Regression is a fundamental supervised learning algorithm used to model the relationship between a dependent variable and one or more independent variables. It predicts continuous values by fitting a straight line that best represents the data.



- It assumes that there is a linear relationship between the input and output
- Uses a best‑fit line to make predictions
- Commonly used in forecasting, trend analysis, and predictive modelling

For example, suppose we want to predict a student’s exam score based on the number of hours studied. As study hours increase, exam scores generally increase as well. Here:

- Independent variable (input): Hours studied because it's the factor we control or observe.
- Dependent variable (output): Exam score because it depends on how many hours were studied.

Linear regression uses the independent variable to predict the dependent variable.



### Best Fit Line

In linear regression, the best-fit line is the straight line that best represents the relationship between the independent variable (input) and the dependent variable (output). The goal is to minimize the difference between the actual data points and the predicted values generated by the model.

1. Goal of best fit line: The goal of linear regression is to find a straight line that minimizes the error (the difference) between the observed data points and the predicted values. This line helps us predict the dependent variable for new, unseen data.



Here Y is called a dependent or target variable and X is called an independent variable also known as the predictor of Y.

2. **Equation of Best-fit line**

For simple linear regression (with one independent variable), the best-fit line is represented by the equation

y = mx+b 

**Where** 

- y is the predicted value(dependent variable)
- x is the input (independent variable)
- m is the slope of the line (how much y changes when x changes)
- b is the intercept (the value of y when x= 0)

**Example**

Suppose we predict salary 
| Experience (Years) | Salary ($1000) |
| ------------------ | -------------- |
| 1                  | 40             |
| 2                  | 50             |
| 3                  | 60             |
| 4                  | 70             |
| 5                  | 80             |

The regression line may become 

Salary = 10 x Experience + 30  

for 6 years:

Salary = 10(6) + 30 = 90

Predicted Salary = $90,000

3. **Minimizing the Error using the least square method**

To determine the best-fit line, linear regression uses the Least Squares Method, which minimizes the difference between actual and predicted values. These differences are called residuals. These differences are called residuals.

The formal for residuals is : 

residual = actual value - predicted value #in simple term 

For Example

- Actual house price = $250,000
- Predicted house price = $240,000

Residual:

e=250000−240000=10000

Residual = 10,000

The model underestimated the house price by $10,000.

** List sqauare method ** - The Least Squares Method is used to find the best-fitting line or curve for a set of data points.

The least squares method minimizes the sum of the squared residuals

formula :

$$
\min \sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

It can be applied in many situations, not only linear regression.

**Ordinary List Square**

OLS is the standard version of the Least Squares Method specifically used for linear regression.

OLS finds the values of: 


- $\beta_0$ (intercept)
- $\beta_1$ (slope)

that minimize the squared errors.

The linear regression equation is:

$$
\hat{y} = \beta_0 + \beta_1 x
$$

OLS finds the best values of $\beta_0$ and $\beta_1$.

Example : 

| Hours Studied (x) | Exam Score (y) |
| ----------------- | -------------- |
| 1                 | 50             |
| 2                 | 60             |
| 3                 | 70             |

We want to find the best-fit regression line:

$$
\hat{y} = \beta_0 + \beta_1x
$$

Different lines can be created:

**Line 1:**

$$
\hat{y} = 40 + 10x
$$

**Line 2:**

$$
\hat{y} = 45 + 8x
$$

For each line, we calculate the Sum of Squared Errors (SSE):

$$
SSE = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

The line with the lowest SSE is selected as the best-fit regression line.

SSE (Sum of Squared Errors) measures how far the predicted values from a regression model are from the actual values.

It is the total of all squared residuals.


This process is called **Ordinary Least Squares (OLS)**.

**Assumption of OLS** 

For OLS regression to provide valid estimates, several assumptions must be met:

- Independence of Observations: Each observation must be independent of others. For example, the height of a person does not depend on the height of the person measured just before them.
- Homoscedasticity: The variance of the residuals (errors) should be constant across all levels of the independent variables. If the variance of errors increases or decreases with the values of the independent variables, the model might not be reliable.
- Normality of Residuals: The residuals should follow a normal distribution. This assumption is necessary for valid hypothesis testing and confidence intervals.
- Linearity: The relationship between the independent and dependent variables should be linear. If the relationship is not linear, OLS will not provide accurate predictions.
- No Multicollinearity: The independent variables should not be highly correlated with each other. If they are, it becomes difficult to separate their individual effects on the dependent variable.

4. **Interpretation of the Best-Fit Line**

- Slope (m): The slope indicates how much the dependent variable changes for every one-unit increase in the independent variable. For example, if the slope is 5, then y increases by 5 units for every 1-unit increase in x.
- Intercept (b): The intercept represents the predicted value of y when x = 0. It’s the point where the line crosses the y-axis.

**Hypothesis Function** 

The hypothesis function is the equation used by a linear regression model to make predictions.

It represents the relationship between input features and the target variable.

For simple linear regression:

$$
\hat{y} = \beta_0 + \beta_1x
$$

Where:

- $\hat{y}$ = Predicted value of dependent variable
- $x$ = Independent variable (feature)
- $\beta_0$ = Intercept (value of y when x = 0)
- $\beta_1$ = Slope/Coefficient (change in y for one unit change in x)


### Example:

Suppose:

$$
\hat{y}=30000+5000x
$$

For 5 years of experience:

$$
\hat{y}=30000+(5000 \times 5)
$$

$$
\hat{y}=55000
$$

Predicted salary = $55,000

**Cost Function** 

The cost function measures how much error exists between actual and predicted values.

The goal of linear regression is to minimize this error and find the best-fit line.

The most common cost function is Mean Squared Error (MSE):

$$
J(\theta)=\frac{1}{n}\sum_{i=1}^{n}(\hat{y_i}-y_i)^2
$$

Where:

- $J(\theta)$ = Cost function
- $n$ = Number of observations
- $\hat{y_i}$ = Predicted value
- $y_i$ = Actual value






### Gradient Descent

Gradient Descent is an optimization technique used to train a linear regression model by minimizing the prediction error. It works by starting with random model parameters and repeatedly adjusting them to reduce the difference between predicted and actual values.



How it works:

- Start with random values for slope and intercept.
- Calculate the error between predicted and actual values.
- Find how much each parameter contributes to the error (gradient).
- Update the parameters in the direction that reduces the error.
- Repeat until the error is as small as possible.

In simple words:

**The gradient tells us how to adjust parameters ($\beta_0$ and $\beta_1$) to minimize the error.**

---

#### Why Do We Use Gradient Function?

In linear regression, we want to find the best values of:

- $\beta_0$ = Intercept
- $\beta_1$ = Slope

that minimize the cost function.

The cost function has a curve, and the gradient helps us move toward the lowest point (minimum cost).

---

#### Gradient Descent Formula

$$
\theta = \theta - \alpha \frac{\partial J(\theta)}{\partial \theta}
$$

Where:

- $\theta$ = Model parameters ($\beta_0, \beta_1$)
- $\alpha$ = Learning rate (step size)
- $J(\theta)$ = Cost function
- $\frac{\partial J(\theta)}{\partial \theta}$ = Gradient of the cost function

---

#### How Gradient Descent Works

Steps:

1. Initialize parameters with random values.
2. Calculate the cost function.
3. Calculate the gradient.
4. Update parameters using the gradient formula.
5. Repeat until the cost becomes minimum.

---

#### Example Calculation

Assume:

$$
\beta_1 = 10
$$

Gradient:

$$
\frac{\partial J}{\partial \beta_1}=4
$$

Learning rate:

$$
\alpha=0.1
$$

Update parameter:

$$
\beta_1 = 10-(0.1 \times 4)
$$

$$
\beta_1=9.6
$$

The parameter moves closer to the value that minimizes the cost.






### Global Minimum and Local Minimum

#### Global Minimum

A **global minimum** is the lowest point of the entire cost function.

It represents the point where the model has the **smallest possible error**.

In machine learning, finding the global minimum means finding the best values of model parameters that minimize the cost function.

Example:

$$
\theta^* = \arg\min J(\theta)
$$

Where:

- $\theta^*$ = Optimal model parameters
- $J(\theta)$ = Cost function

The goal of optimization algorithms like Gradient Descent is to reach the global minimum.


#### Local Minimum

A **local minimum** is a point where the cost function is lower than all nearby points, but it is not necessarily the lowest point in the entire function.

The model may stop at a local minimum even though a better solution exists elsewhere.

#### Difference Between Global Minimum and Local Minimum

| Global Minimum | Local Minimum |
|---|---|
| Lowest point of the entire cost function | Lowest point in a small region |
| Gives the best possible solution | May not be the best solution |
| Has the minimum overall error | Has minimum error only nearby |
| Ideal goal of optimization | Can cause algorithms to stop early |



#### Example in Gradient Descent

Gradient Descent tries to reduce the cost function by moving downhill.

The update formula is:

$$
\theta = \theta - \alpha\frac{\partial J(\theta)}{\partial\theta}
$$

Where:

- $\theta$ = Model parameters
- $\alpha$ = Learning rate
- $J(\theta)$ = Cost function
- $\frac{\partial J(\theta)}{\partial\theta}$ = Gradient

The algorithm updates parameters until it reaches a minimum point.

Example: 

Imagine a mountain landscape:

- The **lowest valley on the entire mountain** = Global Minimum
- A **small valley on one side of the mountain** = Local Minimum

Gradient Descent tries to find the lowest valley (global minimum) to minimize the error.



# R² and Adjusted R² in Linear Regression

## What is R²?

R² (Coefficient of Determination) is the statistical measure of the variance of the regression (best fit) line from the actual data points. In a generalized linear regression model the model accuracy ranging from 75 - 95% is considered as a good prediction model, whereas 100% accuracy means overfitting.

A higher R² means the model explains more variation.

Range:

$$
0 \leq R^2 \leq 1
$$

---

## R² Formula

$$
R^2=1-\frac{SSE}{TSS}
$$

Where:

- $SSE$ = Sum of Squared Errors
- $TSS$ = Total Sum of Squares

**R-squared values range from 0 to 1:**

- An R-squared of 0 indicates that the model explains none of the variability of the dependent variable.
- An R-squared of 1 indicates that the model explains all the variability of the dependent variable.

---

## SSE Formula

$$
SSE=\sum(y_i-\hat{y_i})^2
$$


## TSS Formula

$$
TSS=\sum(y_i-\bar{y})^2
$$


---

# Example Calculation

Actual values:

$$
y=[10,20,30]
$$

Predicted values:

$$
\hat{y}=[12,18,28]
$$


SSE:

$$
SSE=(10-12)^2+(20-18)^2+(30-28)^2
$$


$$
SSE=4+4+4=12
$$


Mean:

$$
\bar{y}=20
$$


TSS:

$$
TSS=(10-20)^2+(20-20)^2+(30-20)^2
$$


$$
TSS=100+0+100=200
$$


R²:

$$
R^2=1-\frac{12}{200}
$$


$$
R^2=0.94
$$


The model explains 94% of the variation.

---

# Adjusted R²

Adjusted R² improves R² by adding a penalty for unnecessary features.

Formula:

$$
Adjusted\ R^2=
1-\frac{(1-R^2)(n-1)}
{n-k-1}
$$


Where:

- $n$ = Number of observations
- $k$ = Number of predictors

---

# Key Difference

|R²|Adjusted R²|
|-|-|
|Measures model performance|Measures performance with penalty|
|Increases with more variables|Increases only when variables improve model|
|May overestimate performance|Better for multiple regression|

```python
#Example of R2 and Adjusted R2

import numpy as np 
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

#create s dataset : Example predict salary based on years of experience 

data = {
    "experience": [1, 2, 3, 4, 5],
    "salary": [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)
 
#split data 

x = df[["experience"]]
y = df["salary"]

#train model 

model = LinearRegression()

model.fit(x,y)

#prediction

y_pred = model.predict(x)

#SEE

SSE = np.sum((y-y_pred)**2)

#TSS

y_mean = np.mean(y)

TSS = np.sum((y-y_mean)**2)


# R2

R2 = 1-(SSE/TSS)



# Adjusted R2

n = len(y)

k = x.shape[1]


Adjusted_R2 = (
    1-((1-R2)*(n-1))/(n-k-1)
)


print("SSE:", SSE)

print("TSS:", TSS)

print("R²:", R2)

print("Adjusted R²:", Adjusted_R2)
```
