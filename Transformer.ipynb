# Lets Understand what is LLM first 

**Large Language Model**

LLM (Large Language Model) is an advanced AI system designed to process, understand, and generate human-like text.

LLMs are usually built using deep neural networks and trained on massive amounts of text data. Through training, they learn patterns, relationships, context, and structures of human language.

**Why LLMs are important?**

LLMs represent a major advancement in Artificial Intelligence because they have changed how people interact with information and technology.

They can understand user intent and perform many language-related tasks without requiring a separate model for every task.

Common Applications:

- Language translation
- Code generation and completion
- Text generation
- Text classification
- Question answering
- Text summarization
- Reasoning and problem solving

**Foundational LLMs**

It is large language model trained on huge amounts of data to learn general language patterns and knowledge. These models can perform many tasks without being specifically trained for every individual task.

For example, general LLM may be able to:

- Answer questions
- Summarize documents
- Translate languages
- Generate code
- Write text
- Explain concepts

This ability to perform tasks without direct task-specific training is one of the important features of modern LLMs.

**Emergent Behaviour**

Emergent behavior refers to abilities that an LLM develops as a result of large-scale training, even though the model was not explicitly trained for that particular task.

For example, an LLM trained on large amounts of text may demonstrate abilities such as:

- Basic reasoning
- Translation
- Summarization
- Question answering
- Code generation

These capabilities can emerge from learning patterns in large and diverse datasets.

**Fine Tuning**

Fine-tuning is the process of taking an already trained LLM and training it further on a specific dataset, task, or domain.

A general-purpose LLM may work well for many tasks, but its performance may not be sufficient for a specialized application.

Fine-tuning allows us to customize the model.

Example:

A general LLM can be fine-tuned using thousands of customer-service conversations to create a customer-support chatbot.

Advantages of Fine-Tuning
- Requires less data than training from scratch
- Requires fewer computational resources
- Takes less time than training a new LLM
- Improves performance on specialized tasks
- Allows customization for specific domains
- Training from Scratch vs Fine-Tuning

Training from scratch:

Huge dataset → Train new model → General-purpose LLM

Fine-tuning:

Pre-trained LLM → Additional specialized training → Task-specific LLM


**Prompt Engineering**

Prompt engineering is the process of designing effective instructions or questions to guide an LLM toward the desired output.

It is sometimes described as the art and science of creating effective prompts.

Example:

Poor prompt:

- Marketing

Better prompt:

- Explain digital marketing to a beginner using five simple bullet points and provide one example for each.

The second prompt provides more context and clearer instructions, which can help produce a better response.

Prompt Engineering Can Control:
- What the model should do
- How the model should respond
- The format of the response
- The amount of detail
- The tone and style
- Relevant context

## How LLM works

LLMs are primarily based on the Transformer architecture which enables them to learn long range dependencies and contextual meaning in text. At a high level, they work through

- Input Embeddings: Converting text into numerical vectors.
- Positional Encoding: Adding sequence/order information.
- Self-Attention: Understanding relationships between words in context.
- Feed-Forward Layers: Capturing complex patterns.
- Decoding: Generating responses step by step.
- Multi-Head Attention: Parallel reasoning over multiple relationships.



### Transformer 

**What are Transformers?**


A **Transformer** is a type of deep learning neural network architecture that is designed to understand relationships between elements in sequential data.

Transformers became especially important in **Natural Language Processing (NLP)** because they can understand the relationship between words or tokens in a sentence.

The Transformer architecture was introduced in the 2017 research paper:

> **"Attention Is All You Need"**

The major innovation was the use of the **self-attention mechanism** instead of processing words sequential using recuurent neural networks.



In simple way we can say that : **Text → Tokens → Numbers → Attention → Contextual Representation → Prediction**

**For example** 

> " The cat sat on the mat"

The transformer does not simply look at each word independently. 

It helps the relationship such as:

- "cat" is related to "sat"
- "sat" is related to "on"
- "mat" is related to "on"
- "the" can relate to different nouns depending on context

The key idea is:

> **Attention allows each token to look at other tokens and determine which ones are important.**



# 2. Why we do need Transformer?

Before the transformer was popular, Natural language Processing System commonly used 

- RNNs (Reccurent Neutral Network)- is a type of artificial neural network designed to process sequential data like text, speech, or time series
- LSTMs(Long Short Term Memory)- It is a special kind of computer program (a neural network) that learns from data in a specific order, like words in a sentence or numbers over time. It remembers important facts for a long time and forgets things that do not matter
- GRUs(Gated Recuurent Units)

The above process a sequence step by step:

For example:

>  " I love machine learning"

An RNN process: 

> I - Love - machine - learning 

The problem is that the model has to carry the information from step to step next. 

For long sentence, information from the beginning can become harder to preserve. 

** Problem with RNNs**

1. **Sequential Processing**

RNNs process tokens one at a time. This makes training less parallelizable. 

2. **Long-range dependencies**

Consider: "The student who studied machine learning for several years finally passed the difficult exam."

To understand the sentence, the model may need to connect words that are far apart.

3. ** Slow Training**

Because RNNs process sequence sequentially, they cannot take full advantage of parallel GPU computation

-------
**Transformer Solution**

Transformer use **Self- Attention**. 

Instead of processing: 

Token 1 → Token 2 → Token 3 → Token 4

the Transformer can consider relationships between many tokens simultaneously.

Therefore:

> **Transformers can process relationships between tokens much more efficiently than traditional recurrent architectures.**

# 3. How does Transformer work?

Think of a transformer like a transalation machine with two main workers:




1. The Encorder - " The Reader"

It reads the input sentence("How are you") and convert it into a rich numerical reperesntation- basically capturing the meaning, contect and relationship between all the words. It doesnt produce the final outputl it just build a deep undertanding of the input. 



2. The Decorder - " The Writer"

It takes that understanding the encorder and generate the output one word a time ("¿Cómo", then "estás", then "?"), using both the encoder's representation and what it's already written so far to decide the next word.



**Why "stack" of encoders/decoders?**

Instead of just one encoder and one decoder, the original design stacks 6 of each on top of each other (N=6). Each layer refines the representation a bit more — like passing a draft through multiple editors, each one sharpening the understanding before passing it to the next. All 6 encoder layers have identical structure to each other, and all 6 decoder layers have identical structure to each other (though encoder ≠ decoder structure).



The flow step by step:

1. Input Sentence - Encorder layer 1 - encorder layer 2-... Encorder layer 6 
2. The final encorder's output feeds into every decorder layer 
3. Decorder layer 1 - Decorder layer 2...- Decorder layer 6 generating the output word by word 
4. Final decorder layer- output sentence

What made this big deal : Older models (RNNs) processed words one at a time in sequence, which was slow and made it hard to remember distant context. Transformers instead use self-attention — a mechanism that lets the model look at all words at once and weigh how relevant each word is to every other word, regardless of distance. That's why it doesn't need RNNs or convolutions at all.

## **Encorder Workflow**

The primary feature of encorder is to transform the input tokenz into contextualized representation. Unlike earlier models that procces takes independently, the transformer encorder captures the context of each token with respect to the entire sequence. 

The role of encorder are:
- Acts as the first major block in the transformer model
- Takes input embeddings and generates representations
- Each encoder layer applies multi-head self-attention and feed-forward networks
- In machine translation, the encoder processes the source language sentence (e.g., "Hello world") and creates rich representations that capture the meaning, context, and relationships between words, which can then be used by a decoder to generate the target language translation.



1. **Step 1: Input Embedding**



--------------------

Embedding happens in the botton most encorder.Words get converted into number vectors (size 512 each). "How are you?" becomes 3 vectors, one per word — capturing rough meaning. This only happens at the very bottom encoder; higher encoders just take the previous encoder's output as their input.

2. **Step 2: Postional Encoding**

Since there's no RNN reading words in order, the model needs another way to know word order. So a special position vector (made from sine/cosine wave patterns) is added to each word's embedding. Now each vector carries both "what the word means" and "where it sits in the sentence."


---------------------------------------------

3. **Step 3: Stack of Encorder layer**

The transformer consist of 6 identical layers in the orginial mode. Each layer has 
- Multi head Self Attention: captures relationship between words
- Feed forward Neural Network- Process the information further 
- Residual Connections + Layer Normalization - Improve learning and stability 

The encorder converts the input sentence into reacg contextual representation 



-----------

**Step 3.1 Multi Head Attention Mechanism**

This multi head attention mechanism utilized a specialized attention mechanism know as "**Self Attention**". It is the mechanism that allows each word/token in sequence to understand its relationship with other words in same seuqence. 

For example:

"How are you?"

When processing "are" the model can pay attention to "you" because they are strongly related. 

It helps the model to understand the contect and meaning of each word based on the entire input sequence. 

**Query, Value and Key**

Self attention used three important components and for every token the transformer create Q, K and V vector. 

- **Query**(Q)- represents what information the current token is looking for. (**What information do I need?**)
- **Key**(K)- reperesnts the information associated with each token. It helps determine whether a particular token is relevant to the Query.(**What infornation do I have?**)
- **Value**(V)- contains the actual information that will be used to construct the attention output.(**What information should i provide?**)

**Steps of Self Attention**

**<u>Step 1: Create Query, Key and Value</u>**

The input embedding are passed through three different linear layers to create:

$$
Q = XW_Q
$$

$$
K = XW_K
$$

$$
V = XW_V
$$

Where:

- $X$ = Input embeddings
- $W_Q$ = Query weight matrix
- $W_K$ = Key weight matrix
- $W_V$ = Value weight matrix

**<u>Step 2: Calculate Attention Scores</u>**

The Query matrix is multiplied by the transpose of the Key matrix:

$$
QK^T
$$

This is a **dot product** that measures the relationship between tokens.

**<u>Step 3: Scale the Attention Scores</u>**

After calulating the attention scores  then the cores are divided by the square root of the key dimension:

$$
\boxed{
\frac{QK^T}{\sqrt{d_k}}
}
$$

Where:

- $Q$ = Query matrix
- $K$ = Key matrix
- $K^T$ = Transpose of Key matrix
- $d_k$ = Dimension of the Key vectors

**Why do we scale?**: When the dimension of the vector is large, the dot product values can become very large and Large value can cause the **Softmax Function** to produce extremly small or large probabilities, which can lead to unstable gradients. 
- Keeps the value at a reasonable range 
- Make Softmax more stable 
- Improve gradient stability 
- Make model training more stable 

**<u>Step 4: Apply Softmax</u>**

After scaling the attention scores, the **Softmax function** is applied.

$$
\boxed{
Attention\ Weights =
softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)
}
$$

Softmax converts the attention scores into values between **0 and 1**.

These values are called **attention weights**.

The attention weights represent how much attention each token should receive.

**Example**

| Word | Attention Weight |
|---|---:|
| The | 0.05 |
| Cat | 0.60 |
| Is | 0.10 |
| Sleeping | 0.25 |

The weights add up to:

$$
0.05 + 0.60 + 0.10 + 0.25 = 1
$$

A **higher attention weight** means that the model considers that word more important for the current token.

**Why Softmax is used?**

Softmax:

- Converts scores into values between 0 and 1
- Makes the scores easier to interpret as weights
- Gives higher importance to larger scores
- Makes the weights sum to 1

**<u>Step 5: Multiply Attention Weights by the Value</u>**

The attention weights are multiplied by the **Value matrix ($V$)**.

$$
\boxed{
Attention\ Output =
Attention\ Weights \times V
}
$$

Therefore:

$$
Attention\ Output =
softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
$$

The Values with higher attention weights contribute more to the final output.

The Values with lower attention weights contribute less.

**Complete Self Attention Formula**


$$
\boxed{
Attention(Q,K,V)
=
softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
}
$$

**Formula Breakdown**

| Part | Meaning |
|---|---|
| $QK^T$ | Calculates similarity between Queries and Keys |
| $\sqrt{d_k}$ | Scales the attention scores |
| Softmax | Converts scores into attention weights |
| $V$ | Provides the information used to create the output |

**Multi head attention**

It is an extenstion of self-attention. Instead of performing self- attention only once, the Transformer perfoms self-attention multiple times in parallel. 

Each indvidual attention mechanism is called attention head. Each head can learn different relationships between words.

---------
**Step 3.2: Add & Norm (residual connection)**

Each sub-layer in the Transformer encoder, such as Multi-Head Attention and the Feed-Forward Neural Network, uses two important techniques:

- Residual Connection: The original input is added to the sub-layer's output.

         Output=Input+Sublayer(Input)

This helps prevent the vanishing gradient problem and makes it easier to train deeper networks.

- Layer Normalization: The result is normalized to make training more stable and efficient.

--------
**Step 3.3: Feed-Forward Neural Network**

After **Multi-Head Attention**, the output is passed through a **Feed-Forward Neural Network (FFN)** to further process and refine the information.

The FFN consists of:

1. **First Linear Layer**
2. **ReLU Activation**
3. **Second Linear Layer**
4. **Residual Connection**
5. **Layer Normalization**

**Process**



--------
**Step 4: Encorder output**

The output of the final encoder layer is a set of **context-rich vectors** representing the input sequence.

This output is passed to the **decoder**, which uses it to focus on the relevant parts of the input when generating the output.

**Key Points**

- The encoder can contain **N stacked layers**.
- Each layer learns different aspects of the input.
- More layers can help the model learn **richer and more complex representations**.
- The final encoder output provides contextual information to the decoder.

## Decorder Workflow

The decorder is the part of a Transformer responsible for generating the output sequence. For example, in machine translation, the encorder reads the Englishb sentence, while the decorder generates the corresponsing sentence one word at a time. 

The decorder uses the information produced by the encorder along with the words it has already generated. 



---------

**<u>Step 1 : Output Embedding</u>**

The decorder starts with start token, such as <start>. The output token are converted into numerical vectors such as embedding layer. 

For example: I love machine learning - is converted into vectors that the neural network can process 

-------------------------

**<u>Step 2 : Positional Encoding</u>**

The decorder needs to know the position of each word in the sequence. Positional encoding adds this positional information to the words embeddings.

For example: 

I → position 1 

love → position 2

machine → position 3

This helps the model understand the order of words.

**<u>Step 3 : Stack of Decoded Layers</u>**

**1. Masked Self Attention**

The first attention mechanism in each decorder layer is called masked self - attention. 

The decorder can look at the words it has already generated but it cannot look at future words. 

For example, when predicting:

"I love ___"

the decoder can use "I" and "love", but it cannot see the correct future word.

This prevents the model from cheating during the training and make the decorder work properly during generation. 

**2. Cross Attention** 

The second attention mechanism is called encorder-decorder attention or cross- attention. 

Here, the decorder looks at the encorder's output to determine which parts of the input are important 

For example, when translating:

"I love cats"

the decorder may focus on "cats" when generating the corresponding French Word 

In cross-attention:

- Queries come from the decoder.
- Keys come from the encoder.
- Values come from the encoder.

This allows the decoder to combine information from the input sentence with the words it is currently generating.

**3.Feed Forward Network**
After attention, the information passes through a feed-forward neural network.

It processes the information independently at each position and helps the model learn more complex patterns.

----------

**<u>Step 4 : Linear Layer and Softmax</u>**

The decoder's final output passes through a linear layer, which produces a score for every word in the vocabulary.

For example, with a vocabulary of 1,000 words, the linear layer produces 1,000 scores.

The softmax function converts these scores into probabilities between 0 and 1. The word with the highest probability is selected as the next predicted word.

**Normalization and Residual Connections**

Each sub-layer (masked self-attention, encoder-decoder attention, feed-forward network) is followed by a normalization step, and each also includes a residual connection around it.

------------

**<u>Output of the decorder</u>**

The decoder uses a linear layer and softmax to predict the next word. The predicted word is added back as input, and this process continues until an end token is generated.

The decoder can have N layers, where each layer builds on the previous one. Multiple layers and attention heads help the model learn different patterns and relationships, improving prediction accuracy.

## Real life Transformer Models 

Transformers are the foundation of many modern NLP and generative AI models. Different organizations have developed Transformer-based models for different purposes.

**1. BERT**

(Bidirectional Encorder Representations from Transformers) was introduced by Google in 2018. 

The key idea behind BERT is bidirectional understanding. Instead of reading a sentence only from left to right, BERT considers the words before and after a word simultaneously.

For example:

"The bank is near the river."

BERT can use the surrounding words, such as "river," to understand that bank refers to the side of a river rather than a financial institution.

**Key Points**

- Based primarily on the Transformer encoder.
- Uses bidirectional context.
- Pre-trained on large amounts of text, including Wikipedia.
- Performs well on tasks such as:
     - Question answering
     - Text classification
     - Sentence understanding
     - Named entity recognition
     - Sentiment analysis

**2. LaMda**

LaMDA (Language Model for Dialogue Applications) was developed by Google and introduced in 2021.

It was specifically designed for conversation and dialogue.

Unlike models focused mainly on understanding individual sentences, LaMDA was designed to maintain a conversation and generate responses that are relevant to the ongoing discussion

**Key Points**

- Useful for : Chatbots, Virtual Assistants and Inteactive AI applications
- Designed specifically for dialogue
- Focuses on conversational context 

**3. GPT and ChatGPT**

GPT (Generative Pre-trained Transformer) models were developed by OpenAI.

GPT models are primarily designed for generating text.

The progression started with:

- GPT-1 — 2018
- GPT-2 — 2019
- GPT-3 — 2020
- GPT-4 and later generations

GPT models use the Transformer architecture to predict the next token based on the previous context.

For example:

"Machine learning is a type of..."

The model may predict:

"artificial intelligence"

This next-token prediction process is repeated to generate longer text.

GPT models can perform tasks such as:

- Text generation
- Translation
- Summarization
- Question answering
- Coding
- Creative writing
- Reasoning

ChatGPT is a conversational AI system built around GPT-family models and optimized to interact with users through dialogue.

**4. Claude**

Claude is a family of AI assistants developed by Anthropic.

Claude is based on Transformer technology and is designed with a strong emphasis on helpfulness, safety, and following instructions.

Anthropic introduced Constitutional AI, an approach that uses a set of principles to help guide model behavior.

Claude models have been designed for tasks such as:

- Question answering
- Writing
- Coding
- Reasoning
- Summarization
- Working with long documents
- Following complex instructions

Claude has evolved through several generations, including Claude 3 and Claude 4 families.






## Modern Transformer Variants


Since the original Transformer architecture was introduced in 2017, many new variants have been developed to improve efficiency and expand Transformers to new types of data.

1. Vision Transformers (ViT)

- Apply Transformers to **images**.
- Divide an image into small **patches** and treat each patch as a token.
- Use self-attention to understand relationships between image patches.
- Used for:
  - Image classification
  - Object detection
  - Image generation

**Remember:** ViT = Transformer for images.

2. Multimodal Transformers

- Process multiple types of data, such as:
  - Text
  - Images
  - Audio
  - Video
- Allow the model to understand relationships between different modalities.

**Remember:** Multimodal = Text + Image + Audio + Video.

3. Efficient Transformers

Efficient Transformers are designed to reduce the computational and memory requirements of standard self-attention.

- **Flash Attention:** Makes attention faster and more memory-efficient.
- **RoPE:** Provides positional information to tokens.
- **Linear Attention:** Helps process very long sequences more efficiently.
- **Mixture of Experts (MoE):** Activates only a subset of model parameters for each input.

**Difference Between Transformer, RNN, and CNN**

| Feature | RNN | CNN | Transformer |

### Limitations of Transformers

Despite their powerful performance, Transformers have several limitations:

1. **High Computational Cost**
   - Self-attention requires significant computational resources, especially for long sequences.

2. **Quadratic Complexity**
   - Standard self-attention has $O(n^2)$ complexity with respect to sequence length.
   - Longer sequences require much more computation and memory.

3. **High Memory Requirements**
   - Large Transformer models require large amounts of GPU memory.

4. **Large Data Requirements**
   - Transformers generally require large datasets to achieve strong performance.

5. **High Training Cost**
   - Training large Transformer models can be expensive and time-consuming.

6. **Hallucinations**
   - Transformers can generate information that sounds correct but is factually incorrect.

7. **Position Encoding**
   - Transformers need positional information to understand the order of tokens.

8. **Energy Consumption**
   - Training and running large Transformer models can consume significant amounts of electricity.
