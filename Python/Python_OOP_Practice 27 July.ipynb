_repr_() Method - method provides a detailed or developer friendly representation of object 

It is useful for 

Debugging | Understanding the object | Viewing object inside collections | Recreating an object when possible 


```python
class Student:
    def __init__(self, name, course):
        self.name = name 
        self.course = course
    
    def __repr__(self):
        return f"Student(name='{self.name}', course='{self.course}')"

student1 = Student("Sadikshya", "Python")

print(repr(student1))
        
```

```python
#_repr with list of an objects

class Product:
    def __init__(self, name , price):
        self.name = name 
        self.price = price 
    
    def __repr__(self):
        return "Product(name'{self.name}', price='{self.price})"
    
product1 = Product ("Lipstick", 20)
product2 = Product ("Foundation", 35)

products = [product1, product2]

print(products)    
```

```python
#Example using str and repr 

class Student:
    def __init__(self, name, course):
        self.name = name 
        self.course = course
    def __str__(self):
        return f"{self.name} is studying {self.course}"
    def __repr__(self):
        return f"Student(name = '{self.name}',course ='{self.course}')"

student1 = Student("Sadikshya","Python")

print(student1)
print(repr(student1))
        
```

Method Overiding 

Method overriding happens when a child class defines a method with the same name as a method in its parent class.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


dog = Dog()

dog.sound()
```

```python
#using Supper()- function allows a child class to access methods from its parent class.
#calling parent constructor with supper 
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id


student1 = Student("Sadikshya", 101)

print(student1.name)
print(student1.student_id)
```

```python
#method oversiding with supper()

class Person:
    def introduce(self):
        print("Hello, I am a person")


class Student(Person):
    def introduce(self):
        super().introduce()
        print("I am also a student")


student1 = Student()

student1.introduce()
```

Abstraction: Abstraction means hiding unnecessary implementation details and showing only the important features.The user should know what an object does without needing to understand every internal step.

Real-Life Example

When using an ATM, a user can:

- Enter a PIN 

- Check the balance

- Withdraw money

The user does not need to understand the database queries and internal banking processes.

```python
class CoffeeMachine:
    def make_coffee(self):
        self._heat_water()
        self._add_coffee()
        self._pour_coffee()
        print("Coffee is ready")
    
    def _heat_water(self):
        print("Heating Water")
    
    def _add_coffee(self):
        print("Adding Coffee")
        
    def _pour_coffee(self):
        print("Pouring coffee into the cup")

machine = CoffeeMachine()

machine.make_coffee()
    
```

```python
#Bank account Example 

class BankAccount:
    def __init__(self, balance):
        self._balance = balance
        
    def deposit(self, amount):
        if amount > 0:
            self._balance +=amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")
            
    def check_balance(self):
        return self._balance
    
account = BankAccount(1000)

account.deposit(500)

print("Balance:", account.check_balance())
        
```

Protecting your abstraction :

Protecting an abstraction means preventing users of a class from depending on its internal implementation.

The class should provide clear public methods while keeping its internal helper methods and data hidden as much as practical.

Python uses naming conventions to show how an attribute or method should be used

```python
#public members

#A public member has no leading underscore 

class Student:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print(self.name)


student1 = Student("Sadikshya")

print(student1.name)
student1.display_name()
```

```python
#protected members- starts with one underscore 

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self._salary = salary

    def display_salary(self):
        print("Salary:", self._salary)


employee1 = Employee("Riya", 50000)

employee1.display_salary()
```

Inheritance 

Inheritance allows a child class to reuse and extend the attributes and methods of a parent class.

It reduces repeated code.

```python
class Person:
    def speak(self):
        print("Person is speaking")


class Student(Person):
    def study(self):
        print("Student is studying")


student1 = Student()

student1.speak()
student1.study()
```

```python
#Single Inheritance : One child class inherits from one parent class.

class Person:
    def introduce(self):
        print("I am a person")


class Student(Person):
    def study(self):
        print("I am studying")


student = Student()

student.introduce()
student.study()

```

```python
#Multiple Inhertance One child class inherits from more than one parent class.

class Writer:
    def write(self):
        print("Writing content")


class Designer:
    def design(self):
        print("Creating a design")


class ContentCreator(Writer, Designer):
    def publish(self):
        print("Publishing content")


creator = ContentCreator()

creator.write()
creator.design()
creator.publish()
```

```python
#Multilevel Inheritance - A class inherits from another child class, creating multiple levels.

class Person:
    def introduce(self):
        print("I am a person")


class Employee(Person):
    def work(self):
        print("I am working")


class Manager(Employee):
    def manage(self):
        print("I am managing a team")


manager = Manager()

manager.introduce()
manager.work()
manager.manage()
```

```python
#Hierarchical Inheritance - Multiple child classes inherit from the same parent class.

class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


class Cat(Animal):
    def meow(self):
        print("Cat is meowing")


dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()

```

```python
#hybrid Inheritance- Hybrid inheritance combines two or more types of inheritance.

class Person:
    def introduce(self):
        print("I am a person")


class Student(Person):
    def study(self):
        print("I am studying")


class Employee(Person):
    def work(self):
        print("I am working")


class TeachingAssistant(Student, Employee):
    def assist(self):
        print("I am assisting the instructor")


assistant = TeachingAssistant()

assistant.introduce()
assistant.study()
assistant.work()
assistant.assist()
```
