```python
#finding missing values 

df.isna()

#counting missing values 

df.isna() and sum()

#dropna- remove rows or column containing missing values 

#remove rows with missing values 

df.dropna()

df.dropna(axis=1) #remove columns with missing values 
```

```python
#fillna()-fills missing values instead of deleting them 

df["Age"] = df["Age"].fillna(0)

#fill with the mean 

df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
#fill with meadian 

df["Age"] = df["Age"].fillna(df["Age"].median())
```

Group By - Combine rows with the same category and perfoms calculation on them 

```python
#avergae salary by city 

df.groupby("City")["Salary"].mean()
```

```python
#multiple calulation 
df.groupby("City")["Salary"].agg(["count","mean","min","max"])
```

```python
#Groups using multiple columns 

df.groupby(["City","Age"])["Salary"].mean()
```

```python
#groupby() with describe()

df.groupby("City")["Salary"].describe()
```

```python
#Transpose - Transpose changes rows into columns and columns into rows

df.transpose()

df.T #shortform 

df.groupby("City")["Salary"].describe().T
```

```python
#combining dataframes

employee = pd.DataFrame({
    "Employee_ID":[1, 2, 3],
    "Name":["Maya","Riya","Sita"]
    })
departments = pd.DataFrame({
    "EmployeeID":[1,2,4],
    "Department": ["Marketing", "Finance","IT"]
})
```

```python
#merging two dataframes

merged_df = pd.merge(
    employees,
    departments,
    on="EmployeeID"
)

print(merged_df)
```

```python
#innerjoin- Returns only matching records.
pd.merge(
    employees,
    departments,
    on="EmployeeID",
    how="inner"
)
```

```python
#leftjoin- Returns every row from the left DataFrame and matching rows from the right DataFrame.
pd.merge(
    employees,
    departments,
    on="EmployeeID",
    how="left"
)
```

```python
#right join-Returns every row from the right DataFrame and matching rows from the left DataFrame

pd.merge(
    employees,
    departments,
    on="EmployeeID",
    how="right"
)
```

```python
#concatenation 
#concat()- Concatenation places DataFrames together.

df1 = pd.DataFrame({
    "Name": ["Maya", "Riya"],
    "Age": [22, 25]
})

df2 = pd.DataFrame({
    "Name": ["Sita", "Anita"],
    "Age": [24, 28]
})

combined = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(combined)
```
