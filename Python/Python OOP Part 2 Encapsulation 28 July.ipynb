Encapsulation in Python is an object-oriented programming (OOP) concept that bundles data (attributes) and the methods (functions) that act on that data into a single unit (a class) while restricting direct access to some of the object's components. This prevents accidental modification of data and hides internal implementation details.

- The idea of encapsulation is to bind the data members and methods into a single unit.
- Helps in better maintainability, readability and usability as we do not need to explicitly pass data members to member methods
- Helps maintain data integrity by allowing validation and control over the values assigned to variables.
- It helps in achieving abstraction. A class can hide the implementation part and discloses only the functionalities required by other classes which allows later changes to representations or implementations without impacting the codes that uses this class.


```python
#example of encapsulation 

class Employee:
    def __init__(self, name, salary):
        self.name = name #public attribute 
        self.__salary = salary #private attribute 
        
emp = Employee ("Sadikshya", 50000)
print(emp.name)
print(emp.__salary)


#self.name = name: Public attribute, can be accessed directly.
#self.__salary = salary: Private attribute, cannot be accessed directly.
#print(emp.name): Prints "Fedrick" because name is public.
#print(emp.__salary): Raises an error because __salary is private and hidden.
```

Composition : This refers to extending a class using the Composite Design Pattern or general object composition. Instead of breaking encapsulation by using inheritance (extends), you encapsulate an instance of another object inside your class to securely extend its features.


Why composition is useful
- Separates a large program into smaller classes
- Allows objects to be reused
- Reduces unnecessary inheritance
- Makes code easier to test and update
- Represents real-world relationships clearly

```python
class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self, brand):
        self.brand = brand
        self.engine = Engine()

    def start_car(self):
        print(self.brand, "is starting")
        self.engine.start()
        
car = Car("BMW")
car.start_car()
```

```python
#another example 

class Author:
    def __init__(self,name):
        self.name = name 
        
class Book:
    def __init__(self, title, author):
        self.title = title 
        self.author = author 
        
    def display_information(self):
        print("Book:", self.title)
        print("Author:", self.author.name)


        
        
```

```python
author = Author("George Orwell")
book = Book("1984", author)

book.display_information()
```

```python
class Text:
    def __init__(self, content):
        self.content = content

    def render(self):
        return self.content
```

```python
class BoldWrapper:
    def __init__(self, text_object):
        self.text_object = text_object

    def render(self):
        return f"<b>{self.text_object.render()}</b>"
```

```python
class ItalicWrapper:
    def __init__(self, text_object):
        self.text_object = text_object

    def render(self):
        return f"<i>{self.text_object.render()}</i>"
```

```python
text = Text("Hello, Python!")

bold_text = BoldWrapper(text)
print(bold_text.render())

italic_text = ItalicWrapper(text)
print(italic_text.render())
```

```python
text = Text("Hello, Python!")

bold_and_italic_text = ItalicWrapper(BoldWrapper(text))

print(bold_and_italic_text.render())
```

Polymorphism in Python 

Polymorphism means "many forms" and allows the same method, function or operator to behave differently depending on the object or data it works with. This flexibility helps create more reusable, maintainable and scalable code.

Types of Polymorphism 

- Compile time polymorphism 
- Runtime polymorphism 

```python
class Dog:
    def speak(self):
        return "Woof!"


class Cat:
    def speak(self):
        return "Meow!"


class Cow:
    def speak(self):
        return "Moo!"
```

```python
animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.speak())
```

```python
#polyphorsim through methor overiding- A child class can replace a method inherited from its parent class. This is called method overriding.

class Employee:
    def calculate_salary(self):
        return 3000


class Manager(Employee):
    def calculate_salary(self):
        return 5000


class Intern(Employee):
    def calculate_salary(self):
        return 1500
```

```python
employees = [Employee(), Manager(), Intern()]

for employee in employees:
    print(employee.calculate_salary())
```

Duck typing means Python focuses on what an object can do, rather than what class or type it belongs to.

It comes from the saying:

If it walks like a duck and quacks like a duck, treat it like a duck.

In Python, if an object has the required method, it can be used—even if it does not inherit from a particular class.

```python
class Duck:
    def move(self):
        return "The duck is walking"


class Person:
    def move(self):
        return "The person is walking"


class Robot:
    def move(self):
        return "The robot is moving"
```

```python
def make_it_move(object):
    print(object.move())
```

```python
make_it_move(Duck())
make_it_move(Person())
make_it_move(Robot())
```

Instance Variable : An instance variable belongs to one specific object. Different objects can have different values.

Instance variables are commonly created inside __init__() using self.

```python
class Student:
    school_name = "Wright State University"

    def __init__(self, name, score):
        self.name = name      # Instance variable
        self.score = score    # Instance variable
```

```python
student1 = Student("Sadi", 90)
student2 = Student("Alex", 82)

print(student1.name, student1.score)
print(student2.name, student2.score)
```

```python
class Employee:
    company = "ABC Company"  # Class variable

    def __init__(self, name, salary):
        self.name = name      # Instance variable
        self.salary = salary  # Instance variable


employee1 = Employee("Sam", 50000)
employee2 = Employee("Maya", 60000)

print(employee1.company)
print(employee2.company)

print(employee1.name, employee1.salary)
print(employee2.name, employee2.salary)
```

Instance Method : An instance method works with the data belonging to a particular object.

- Its first parameter is self.
- It can access and change instance variables.
- It can also access class variables.
- It is normally called using an object.

```python
class Student:
    school = "Python Academy"

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def display_details(self):
        print("Name:", self.name)
        print("Score:", self.score)
        print("School:", self.school)
```

```python
#calling instance method 

student1 = Student("Sadikshya", 90)
student1.display_details()
```

Class Method: A class method works with the class itself rather than one specific object.

- It uses the @classmethod decorator.
- Its first parameter is cls.
- It can access and change class variables.
- It is normally called using the class name.

```python
class Student:
    school = "Python Academy"

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school
```

```python
Student.change_school("Data Science Academy")

print(Student.school)
```

Static Method: A static method is a utility function placed inside a class because it is related to that class.

- It uses the @staticmethod decorator.
- It does not use self.
- It does not use cls.
- It cannot directly access instance or class data.
- It is normally called using the class name.

```python
class Student:
    @staticmethod
    def is_passing(score):
        return score >= 60
```

```python
print(Student.is_passing(85))
print(Student.is_passing(45))
```

```python
class Employee:
    company = "ABC Company"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    # Instance method
    def display_employee(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Company:", self.company)

    # Class method
    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

    # Static method
    @staticmethod
    def is_valid_salary(salary):
        return salary > 0
```

```python
employee1 = Employee("Maya", 50000)

employee1.display_employee()

Employee.change_company("XYZ Company")

print(Employee.is_valid_salary(50000))
print(Employee.is_valid_salary(-100))
```



A decorator is a function that adds extra behavior to another function or method without changing its original code.

A decorator uses the @ symbol.

```python
def welcome_decorator(function):
    def wrapper():
        print("Welcome!")
        function()
        print("Thank you!")

    return wrapper


@welcome_decorator
def say_hello():
    print("Hello, Sadikshya!")


say_hello()
```

```python
#Decorator with arguments 

#Use *args and **kwargs so the decorator can accept different function arguments.

def log_call(function):
    def wrapper(*args, **kwargs):
        print("The function is starting")

        result = function(*args, **kwargs)

        print("The function has finished")
        return result

    return wrapper


@log_call
def add(a, b):
    return a + b


answer = add(5, 3)
print(answer)
```

Dataclass: A data class is a class mainly used to store data.

Python’s @dataclass decorator automatically creates common methods such as:

__init__()
__repr__()
__eq__()

It is available from Python’s dataclasses module.

```python
#Regular class 

class Product:
    def __init__(self, name, brand, price):
        self.name = name
        self.brand = brand
        self.price = price

    def __repr__(self):
        return (
            f"Product(name={self.name}, "
            f"brand={self.brand}, price={self.price})"
        )
```

```python
#using dataclass

from dataclasses import dataclass


@dataclass
class Product:
    name: str
    brand: str
    price: float
```

```python
product = Product("Face Cream", "Glow Beauty", 24.99)

print(product)
```

How long should a class be?

There is no fixed rule such as “a class must be fewer than 50 lines.”

A class should be:

Long enough to complete one responsibility
Small enough to understand easily
Focused on one purpose
Easy to test and modify

The number of responsibilities matters more than the number of lines.

A 100-line class with one clear responsibility may be acceptable. A 30-line class performing five unrelated jobs may need to be divided.

Single Responsibility Principle

The Single Responsibility Principle, or SRP, states:

A class should have one responsibility and one reason to change.

This does not mean that a class can only have one method. It can have several methods, as long as they all support the same responsibility.

```python
#class with too many responsibilities

class ProductManager:
    def calculate_discount(self, price, discount):
        return price - discount

    def save_to_database(self, product):
        print("Saving product to database")

    def send_email(self, message):
        print("Sending email:", message)

    def create_report(self):
        print("Creating sales report")
```

```python
#imporved design using SRP

class PriceCalculator:
    def calculate_discount(self, price, discount):
        return price - discount


class ProductRepository:
    def save(self, product):
        print("Saving product to database")


class EmailService:
    def send(self, message):
        print("Sending email:", message)


class ReportGenerator:
    def create(self):
        print("Creating sales report")
```
