# What is Decision Tree ?

A decision tree is a supervised learning algorithm used for both classification and regression tasks. It has a hierarchical tree structure which consists of a root node, branches, internal nodes and leaf nodes. It works like a flowchart that helps in making step-by-step decisions, where:



- Internal nodes represent attribute tests
- Branches represent attribute values
- Leaf nodes represent final decisions or predictions.


It works by repeatedly asking a series of questions about the input features and splitting the data into smaller groups until it reaches a final prediction.

Decision trees are widely used due to their interpretability, flexibility and low preprocessing needs.

For example, if we want to predict whether a customer will churn, the tree might ask:

- Is the customer's contract month-to-month?
- Is the customer's tenure less than 12 months?
- Is the monthly charge high?
- Does the customer have technical support?

The tree continues making decisions until it reaches a final prediction.






## Basic Structure of Decision Tree

A Decision Tree contains several important components:

Root Node

The Root Node is the first and top-level decision in the tree.

It represents the feature that provides the best initial split.

                 Contract Type
                /             \
       Month-to-month        Long-term

Here, Contract Type is the root node.

---- 
Internal Node

An Internal Node represents another decision or question.

                 Contract Type
                /             \
       Month-to-month        Long-term
              |
         Tenure < 12?

Here, Tenure < 12? is an internal node.
----
Branch

A Branch represents the outcome of a decision.

             Tenure < 12?
              /        \
            Yes         No

The two branches are:

- Yes
- No
-----
Leaf Node

A Leaf Node contains the final prediction.

             Tenure < 12?
              /        \
            Yes         No
             |           |
           Churn       No Churn

Churn and No Churn are leaf nodes.

#### Decision tree are useful :

- Make classification predictions
- Make regression predictions
- Capture nonlinear relationships
- Capture interactions between variables
- Handle complex decision boundaries
- Be easily visualized
- Be relatively easy to interpret
- Work without requiring feature scaling

### Types of Decision Tree

2. **Decision Tree For Regression**

### How does Decision Tree Work ?

A Decision Tree works through a process called recursive partitioning.

The basic process is 

- Dataset 
- Find the best feature
- Find the best split 
- Split the dataset
- Evaluate the resulting groups
- Find another best split 
- Split Again 
- Continue recursively 
- Stop when stopping criteria are reached
- Final prediction

Step by Step process 

1. Start with the Entire Dataset

Initially, the Decision Tree contains all training observations.

                 Entire Dataset
                       |
          ┌────────────┴────────────┐
          ↓                         ↓
      Class A                    Class B

The data may be mixed.

2. Step 2: Examine the Features

The algorithm examines the available features.

For example:

- Age
- Income
- Tenure
- Contract
- Monthly Charges
- Payment Method

The algorithm evaluates which feature can create the most useful split.

3. Find the Best Split

The algorithm evaluates possible splits using a criterion such as:

**Classification**

- Gini Impurity
- Entropy
- Information Gain
- Log Loss

**Regression**

- Mean Squared Error
- Mean Absolute Error
- Variance Reduction

4. Split the Data

The best split divides the data into smaller groups.

Example:

                     Contract
                    /        \
              Monthly       Long-term

5. Repeat the Process

The algorithm repeats the same process for each child node.

                     Contract
                    /        \
              Monthly       Long-term
                 |
               Tenure
              /     \
           < 12     >= 12

6. Stop the Tree

The tree stops growing when a stopping condition is reached.

Examples:

- Maximum depth reached
- Too few samples remain
- No useful improvement from another split
- Maximum number of leaves reached

7. Make a Prediction

The final prediction is made at the leaf node.

Contract = Monthly-Tenure < 12-Churn

#### What is Splitting ?

Splitting means dividing the dataset into smaller groups based on a feature.

For example:

                Contract
               /        \
          Monthly      Long-term

The original dataset has now been divided into two groups.

The purpose of splitting is to create groups that are as pure as possible.

#### What is Purity ?

Purity describes how similar the observations in a node are with respect to the target variable.

Suppose a node contains:

- 95% Churn
- 5% No Churn

This node is highly pure.

Another node contains:

- 50% Churn
- 50% No Churn

This node is highly mixed.

Therefore:

High Purity----Better Separation

Decision Trees try to create increasingly pure child nodes.

#### How Does the Tree Choose the Best Split?

The Decision Tree does not randomly choose a feature.

It evaluates possible splits and selects the one that provides the best improvement according to the chosen criterion.

For classification, common criteria are:

- Gini Impurity
- Entropy
- Information Gain
- Log Loss

For regression:

- MSE
- MAE
- Variance Reduction

### Gini Impurity 

Gini Impurity measures how mixed the classes are in a node.

The formula is:

$$
Gini = 1 - \sum_{i=1}^{K} p_i^2
$$

Where:

- $K$ = number of classes
- $p_i$ = proportion of observations belonging to class $i$

Properties:

- Lower values indicate cleaner and more homogeneous nodes.
- Nodes become pure when all samples belong to one class.
- Slightly biased toward dominant classes during split selection.

Example of Gini Impurity 

Suppose a node contains:

80% Churn
20% No Churn

Then:

$$
Gini = 1 - (0.8^2 + 0.2^2)
$$

$$
Gini = 1 - (0.64 + 0.04)
$$

$$
Gini = 1 - 0.68
$$

$$
Gini = 0.32
$$

Therefore:

Gini = 0.32

------

**Interpretation of Gini Impurity**

A lower Gini value means greater purity.

Gini = 0- Completely Pure

For binary classification:

Gini = 0-Pure node

Gini = 0.5-Maximum impurity

Maximum impurity occurs when:

50% Class A
50% Class B

------

**Weighted Gini Impurity**

When evaluating a split, we consider the impurity of both child nodes.

The weighted impurity is:

$$
\frac{N_L}{N}Gini_L
+
\frac{N_R}{N}Gini_R
$$

Where:

- $N$ = total number of observations
- $N_L$ = observations in left child
- $N_R$ = observations in right child
- $Gini_L$ = Gini impurity of left child
- $Gini_R$ = Gini impurity of right child

The tree prefers the split with the lowest weighted Gini impurity.



#### Entropy 

Entropy is another measure of impurity or uncertainty.

The formula is:

$$
Entropy =
-\sum_{i=1}^{K}p_i\log_2(p_i)
$$

Where:

- $K$ = number of classes
- $p_i$ = probability/proportion of class $i$

Properties:

- Zero entropy corresponds to perfectly pure splits.
- Sensitive to small fluctuations across class ratios.
- Often yields balanced splits with meaningful boundaries.


Example 

Suppose:

- 100% Churn
- 0% No Churn

Then:

$$
Entropy = 0
$$

There is no uncertainty.

Now suppose:

- 50% Churn
- 50% No Churn

Then:

$$
Entropy = 1
$$

for binary classification.

Therefore:

Low Entropy ↓ More Pure High Entropy ↓ More Mixed

### Information Gain 

Information Gain tells us how useful a question (or feature) is for splitting data into groups. It measures how much the uncertainty decreases after the split. A good question will create clearer groups and the feature with the highest Information Gain is chosen to make the decision.

#### Regression Splitting 

For regression, the algorithm attempts to create groups with similar target values.

Common criteria include:

- Mean Squared Error
- Mean Absolute Error
- Variance Reduction

Example:

Group 1:

- $500,000
- $510,000
- $495,000

These values are close together, so the group has relatively low variation.

##### Mean Squared Error 

Mean Squared Error is commonly used for regression trees.

The formula is:

$$
MSE =
\frac{1}{n}
\sum_{i=1}^{n}(y_i-\hat{y})^2
$$

Where:

- $y_i$ = actual value
- $\hat{y}$ = predicted value
- $n$ = number of observations

The Decision Tree tries to create splits that reduce prediction error.

##### Recursive Partitioning

Decision Trees use recursive partitioning.

This means that after the first split, each resulting group can be split again.

Example:

                    Contract
                   /        \
              Monthly       Long-term
                 /              \
              Tenure           Charges
             /     \           /     \
          < 12     >=12     < 70    >=70
            |         |       |        |
          Churn      Stay    Stay     Churn

The same splitting process is repeatedly applied.


##### Stopping Criteria

A Decision Tree must eventually stop growing.

Common stopping criteria include:

1. Maximum Depth
max_depth=5
2. Minimum Samples Required for Splitting
min_samples_split=10
3. Minimum Samples in a Leaf
min_samples_leaf=5
4. Maximum Number of Leaves
max_leaf_nodes=20
5. No Meaningful Improvement

If another split does not improve the model sufficiently, the tree can stop splitting.

### Overfiting In Decision Tree 

A major problem with Decision Trees is overfitting.

If the tree becomes too deep, it can memorize the training dataset.

Example:

- Training Accuracy = 99%
- Testing Accuracy  = 72%

This indicates that the model performs very well on training data but poorly on unseen data.

That is a sign of possible overfitting.

**Why Do Decision Trees Overfit?**

A Decision Tree can continue creating increasingly specific rules.

For example:

Age > 35
   ↓
Income > $70,000
   ↓
Tenure < 8 months
   ↓
Monthly Charges > $85
   ↓
Specific Customer

If the tree becomes extremely deep, it may start learning noise rather than general patterns.

**How to Prevent Overfitting**

We can control overfitting using:

- max_depth
- min_samples_split
- min_samples_leaf
- max_leaf_nodes
- max_features
- ccp_alpha
- Cross-validation

### Pre Purning 

Pre-pruning means restricting the tree while it is being created.

- model = DecisionTreeClassifier( max_depth=5, min_samples_split=10, min_samples_leaf=5 )

The tree is prevented from becoming unnecessarily complex.


### Post Purning 

Post-pruning means allowing the tree to grow and then removing unnecessary branches.

One method in scikit-learn is Cost-Complexity Pruning.

The parameter is:

ccp_alpha

Example:

model = DecisionTreeClassifier(
    ccp_alpha=0.01
)

A larger ccp_alpha generally means stronger pruning.


### Cost Complexity Purning 

Cost-complexity pruning balances:

Prediction Error
       +
Tree Complexity

The goal is to find a tree that provides good predictions without unnecessary complexity.

Conceptually:

$$
Cost =
Error + \alpha \times Complexity
$$

Where:

$\alpha$ controls the penalty for complexity
Higher $\alpha$ encourages simpler trees

## Cross Validation 

Cross-validation is a technique used to check how well a machine learning model performs on unseen data while preventing overfitting. It works by:

- Splitting the dataset into several parts.
- Training the model on some parts and testing it on the remaining part.
- Repeating this resampling process multiple times by choosing different parts of the dataset.
- Averaging the results from each validation step to get the final performance.

Validation Techniques :

1. Holdout Validation :

In Holdout Validation the dataset is split into training and testing sets. Common splits include 70–30, 80–20 or 75–25 depending on the dataset size and problem. Making it simple and quick to apply.

2. LOOCV (Leave One Out Cross Validation)

In this method the model is trained on the entire dataset except for one data point which is used for testing. This process is repeated for each data point in the dataset.

- All data points are used for training, resulting in low bias.
- Testing on a single data point can cause high variance, especially if the point is an outlier.
- It can be very time-consuming for large datasets as it requires one iteration per data point.

3. Stratified Cross-Validation

It is a technique that ensures each fold of the cross-validation process has the same class distribution as the full dataset. This is useful for imbalanced datasets where some classes are underrepresented.

- The dataset is divided into k folds, keeping class proportions consistent in each fold.
- In each iteration, one fold is used for testing and the remaining folds for training.
- This process is repeated k times so that each fold is used once as the test set.
- It helps classification models generalize better by maintaining balanced class representation.

4. K fold cross validation 

K-Fold Cross Validation splits the dataset into k equal-sized folds. The model is trained on k-1 folds and tested on the remaining fold. This process is repeated k times each time using a different fold for testing.

Note: A commonly used value of k is 10, but the choice depends on the dataset size and problem requirements.
