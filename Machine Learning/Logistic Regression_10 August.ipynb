# Logistic Regression 

**Definition** - It is a supervised machine learning algorithm used for classification problems.Unlike linear regression, which predicts continuous values it predicts the probability that an input belongs to a specific class.

- It is used for binary classification where the output can be one of two possible categories such as Yes/No, True or False or 0/1 
- It used a sigmoid function to convert inputs into a probability value between 0 and 1

**Example**
- Predicting whether a customer will buy a product (Yes/No).
- Detecting whether an email is spam or not spam.
- Predicting whether a patient has diabetes (Yes/No).

**Difference between Logistic Regression and Linear Regression**

- Logistic Regression predicts a class and category where as linear regression predicts a contionus value 
- Logistic Regression deals with output probability between 0 and 1 where as Linear regression deals with any numeric value. 
- Logistic Regression function is sigmoid (s shaped function ) where as linear regression is straight line regression. 



## Understanding Sigmoid Function 

Sigmoid is a mathematical function that maps any real-valued number into a value between 0 and 1. Its characteristic "S"-shaped curve makes it particularly useful in scenarios where we need to convert outputs into probabilities. This function is often called the logistic function.

 In logistic regression, we use a threshold value usually 0.5 to decide the class label.
 - If the sigmoid output is same or above the threshold, the input is classified as Class 1.
 - If it is below the threshold, the input is classified as Class 0.

logistic Regression computes a linear combination of input features (y=mx+c) and passes it through a sigmoid function to produce a probability between 0 and 1. This probability is then used to assign the input to a class. 

**Formula**

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

Where:

- **z** = input from the logistic regression equation
- **e** = Euler's number (approximately 2.718)
- **σ(z)** = output probability

**Example**

Suppose the logistic regression model calculates:

$$
z = 2
$$

Then:

$$
\sigma(2) = \frac{1}{1 + e^{-2}}
$$

$$
\sigma(2) \approx 0.88
$$

So, the model predicts an **88% probability**.

For example, if we are predicting **customer churn**:

- Probability = **0.88**
- Threshold = **0.50**
- Prediction = **Churn (1)**

**Why Do We Need the Sigmoid Function?**

A linear equation can produce values such as:

A linear equation can produce values such as:

- -10
- 5
- 20
- 100

These values cannot be directly interpreted as probabilities because a probability must be between **0 and 1**.

The sigmoid function converts these values into probabilities:

| Input (z) | Sigmoid Output |
|-----------|----------------|
| -10 | 0.000 |
| -2 | 0.119 |
| 0 | 0.500 |
| 2 | 0.881 |
| 10 | 1.000 |

### Key Points

- Negative values → probability close to **0**
- 0 → probability of **0.5**
- Positive values → probability close to **1**
- Output is always between **0 and 1**

> **Sigmoid function = converts the model's raw score into a probability between 0 and 1.**

This is why the sigmoid function is important in **Logistic Regression**.

## Types of Logistic Regression 

Logistic Regression is mainly divided into three types based on the **number of classes/categories** in the target variable:

1. **Binary Logistic Regression**
2. **Multinomial Logistic Regression**
3. **Ordinal Logistic Regression**



#### 1. Binary Logistic Regression

Used when the dependent variable has only two possible categories, such as Yes/No, Pass/Fail, or 0/1. It is the most common type and is used for binary classification tasks.
**Examples**

- Customer → Churn / No Churn
- Email → Spam / Not Spam
- Loan → Default / No Default
- Student → Pass / Fail
- Customer → Purchase / No Purchase

#### 2. Multinomial Logistic Regression

Used when there are 3 or more categories that have no natural order.

**Examples:**

- Product choice → Laptop / Phone / Tablet
- Transportation → Car / Bus / Train
- Customer preference → Basic / Standard / Premium (if these are treated as categories without order)

Example:

Predict which product a customer will buy.
- Laptop  → 20%
- Phone   → 60%
- Tablet  → 20%

Prediction- Phone

##### **Softmax function**

The **Softmax function** is a mathematical function mainly used in **multi-class classification**.

It converts the model's **raw scores (logits)** into **probabilities** for each class.

The probabilities produced by Softmax always add up to **1 (100%)*

For a model with multiple classes:

$$
P(y=k) = \frac{e^{z_k}}{\sum_{j=1}^{K} e^{z_j}}
$$

Where:

- $z_k$ = raw score (logit) for class $k$
- $e$ = Euler's number, approximately 2.718
- $K$ = total number of classes
- $P(y=k)$ = probability of class $k$

**Example**

Suppose we want to predict which transportation method a customer will choose:

- Car
- Bus
- Train

The model produces the following raw scores:

| Class | Raw Score |
|-------|----------:|
| Car | 2.0 |
| Bus | 1.0 |
| Train | 0.5 |

These scores are **not probabilities** because they do not represent values between 0 and 1.

Softmax converts them into probabilities:

| Class | Probability |
|-------|------------:|
| Car | 62.9% |
| Bus | 23.1% |
| Train | 14.0% |
| **Total** | **100%** |

The class with the highest probability is selected as the prediction.

**Prediction = Car**

**Difference between Sigmoid and Softmax**

|Sigmoid | Softmax |
|--------|----------|
|Mainly used for binary classification|	Mainly used for multi-class classification|
|Usually focuses on one probability	| Produces probabilities for multiple classes|
|Output is between 0 and 1 | All class probabilities add up to 1|
|Example: Churn / No Churn| Example: Car / Bus / Train|

#### Ordinal Logistic Regression:

 Used when the dependent variable has three or more categories with a natural order, such as Low, Medium, and High. It considers the ranking of categories during prediction.


## Evaluation of classification Model

### Confusion Matrix 

Confusion matrix is a simple table used to measure how well a classification model is performing. It compares the predictions made by the model with the actual results and shows where the model was right or wrong. This helps you understand where the model is making mistakes so you can improve it. It breaks down the predictions into four categories:

- True Positive (TP): The model correctly predicted a positive outcome i.e the actual outcome was positive.
- True Negative (TN): The model correctly predicted a negative outcome i.e the actual outcome was negative.
- False Positive (FP): The model incorrectly predicted a positive outcome i.e the actual outcome was negative. It is also known as a Type I error.
- False Negative (FN): The model incorrectly predicted a negative outcome i.e the actual outcome was positive. It is also known as a Type II error.



#### Metrics Based on Confusion Matrix Data 

##### Accuracy 

It shows how many predictions the model got right out of all the predictions. It gives idea of overall performance but it can be misleading when one class is more dominant over the other.For example a model that predicts the majority class correctly most of the time might have high accuracy but still fail to capture important details about other classes. It can be calculated using the below formula:

Accuracy= TP+TN+FP+FN / TP+TN

- Accuracy works well when the classes are relatively balanced. E.g. 50% churn | 50% not churn 
- Accuracy can be misleading when the dataset is imbalanced.

Suppose:

- 990 customers don't churn
- 10 customers churn

A model predicts everyone as no churn.

It gets:

Accuracy=1000/ 990 =99%

99% sounds excellent, but the model identified zero churners.

That's why we need other metrics.
​
##### Recall 

It measures how how good the model is at predicting positives. It shows the proportion of true positives detected out of all the actual positive instances. High recall is essential when missing positive cases has significant consequences like in medical tests.

It is also called sensitivity or TPR (True positive rate)

Recall= TP+FN/TP

**Advantages of Recall**
- Useful when False Negatives are costly
- Focuses on finding as many positive cases as possible
- Useful for imbalanced datasets

**Limitation of Recall**

- High recall can produce many False Positives
- Recall doesn't tell you whether positive predictions are reliable
- Recall ignores True Negatives

##### Precision 

Precision is the ratio between the True Positives and all the Positives. It shows how many of the “yes” predictions made by the model were actually correct. It helps us reduce wrong “yes” guesses which are called false positives (FP). Precision is calculated as:

Precision = TP/ TP + FP

For example : if most emails are not spam and only a few are then precision helps us see how well the model is finding the spam without making too many mistakes. In such uneven data precision helps measure how correctly the model is picking out the less common group like spam or fraud.

**Advantages of Precision** 

- Useful when False Positives are costly
- Useful when you want reliable positive predictions
- Useful for imbalanced datasets

**Limitation of Precision**

- Precision does not tell you how many positives you missed. A model can have very high precision but very poor recall.
- Precision ignores True Negatives
- High precision can come at the cost of recall
​
**Trade off**

The trade-off between precision and recall happens because your model's classification decision depends on a single threshold, and changing it helps one metric while hurting the other.When you make a model stricter, you eliminate false alarms (boosting precision) but miss real cases (dropping recall). When you make it more lenient, you catch all real cases (boosting recall) but invite false alarms (dropping precision).

**How to manage trade off**
- The F1-Score: This is a single metric that calculates the harmonic mean of precision and recall. It gives you a high score only when both metrics are balanced.
- PR Curve (Precision-Recall Curve): A plot of precision (y-axis) and recall (x-axis) for every possible threshold. You look at this curve to physically pick the threshold that matches your exact business cost for mistakes.

##### F1 Score 

F1 Score is a metric used to evaluate the performance of a classification model. It combines precision and recall into a single value and is especially useful when the dataset has imbalanced classes.

- Combines precision and recall into one metric.
- Useful for imbalanced datasets.
- Higher F1 Score means better performance.

f1 score = 2 X precision x recall/ precision + recall 

The F1 score becomes high only when both precision and recall are high. If either of them decreases significantly, the F1 score will also decrease.

##### Specifcity

Specificity is important to check how well a model can correctly find negative cases. It is useful when making a wrong positive prediction (false positive) causes big problems. For example in email spam filters a false positive might mark an important work email as spam causing you to miss an urgent messages.

Specificity is also also termed as True Negative Rate (TNR) which assess how well the model correctly predicts the absence of a condition and then avoids false alarms.

Specifcity = TN/ TN + FP

**Importance of Specificity**

- Reduces False Alarms
- Essential for Risk Management
- Balances Sensitivity in Imbalanced Datasets
- Improves Trust and Efficiency




### AUC-ROC Curve in Machine Learning

AUC-ROC curve is a graph used to check how well a binary classification model works. It helps us to understand how well the model separates the positive cases like people with a disease from the negative cases like people without the disease at different threshold level. It shows how good the model is at telling the difference between the two classes by plotting:

- True Positive Rate (TPR): How often the model correctly predicts the positive cases also known as Sensitivity or Recall.
- False Positive Rate (FPR): How often the model incorrectly predicts a negative case as positive.
- Specificity: Measures the proportion of actual negatives that the model correctly identifies. It is calculated as 1 - FPR.



#### ROC Curve 

The ROC curve stands for Receiver Operating Characteristics Curve and is an evaluation metric for classification tasks and it is a probability curve that plots sensitivity and specificity. So, we can say that the ROC Curve can also be defined as the evaluation metric that plots the sensitivity against the false positive rate. The ROC curve plots two different parameters given below:

- True positive rate

(TPR) / Sensitivity / Recall (y-axis): Of all actual positives, how many did the model correctly catch?

$$
TPR = \frac{TP}{TP + FN}
$$

- False positive rate

"Of all the *actual* negatives, how many did we *incorrectly* flag as positive?"

$$
FPR = \frac{FP}{FP + TN}
$$

#### AUC Curve

It measures the area under the ROC curve. A higher AUC value indicates better model performance as it suggests a greater ability to distinguish between classes. An AUC value of 1.0 indicates perfect performance while 0.5 suggests it is random guessing.

$$
AUC = \int_0^1 TPR \, d(FPR)
$$

**When to Use AUC-ROC**
AUC-ROC is effective when:

- The dataset is balanced and the model needs to be evaluated across all thresholds.
- False positives and false negatives are of similar importance.

**Why Use ROC/AUC Instead of Accuracy?**

| Metric | Weakness |
|---|---|
| **Accuracy** | Misleading on imbalanced data (predicting "all negative" on a 95%-negative dataset gives 95% accuracy while being useless) |
| **Precision / Recall at one threshold** | Only tells you about *that* threshold — doesn't show overall model quality |
| **ROC / AUC** | Threshold-independent, and less sensitive to class imbalance than accuracy |



```python
import numpy as np
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Example data
y_true = [1, 1, 0, 0, 1]
y_scores = [0.90, 0.70, 0.60, 0.40, 0.30]

fpr, tpr, thresholds = roc_curve(y_true, y_scores)
auc_value = roc_auc_score(y_true, y_scores)

plt.figure(figsize=(6, 6))
plt.plot(fpr, tpr, marker='o', label=f'ROC curve (AUC = {auc_value:.3f})')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Random guess')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()

print(f"AUC: {auc_value:.3f}")
```
